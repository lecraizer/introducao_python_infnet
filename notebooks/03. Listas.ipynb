## 1. O que é uma lista?

Uma **lista** é uma estrutura que permite guardar **vários valores em uma única variável**.

Exemplo:

```python
notas = [7.5, 8.0, 6.5, 9.0]
```

Nesse caso, a variável `notas` guarda vários números ao mesmo tempo.

Listas em Python:

- mantêm a ordem dos elementos
- podem ter valores repetidos
- podem ser alteradas
- podem guardar valores de tipos diferentes

````python
notas = [7.5, 8.0, 6.5, 9.0]
print(notas)
print(type(notas))
````

## 2. Criando listas

Podemos criar listas de várias formas.

### Lista de números
```python
nums = [10, 20, 30]
```

### Lista de strings
```python
nomes = ['Ana', 'Bruno', 'Carlos']
```

### Lista vazia
```python
valores = []
```

### Lista com tipos mistos
```python
dados = ['Luis', 30, True, 1.75]
```

````python
nums = [10, 20, 30]
nomes = ['Ana', 'Bruno', 'Carlos']
vazia = []
dados = ['Luis', 30, True, 1.75]

print(nums)
print(nomes)
print(vazia)
print(dados)
````

## 3. Acessando elementos da lista

Cada elemento possui uma posição chamada **índice**.

Os índices começam em **0**.

Exemplo:

```python
nomes = ['Ana', 'Bruno', 'Carlos']
```

Então:

- `nomes[0]` → `'Ana'`
- `nomes[1]` → `'Bruno'`
- `nomes[2]` → `'Carlos'`

````python
nomes = ['Ana', 'Bruno', 'Carlos']
print(nomes[0])
print(nomes[1])
print(nomes[2])
````

Também podemos usar **índices negativos**:

- `lista[-1]` → último elemento
- `lista[-2]` → penúltimo elemento

````python
nomes = ['Ana', 'Bruno', 'Carlos', 'Daniel']
print(nomes[-1])
print(nomes[-2])
````

## 4. Alterando elementos

As listas são **mutáveis**, ou seja, podem ser alteradas.

Exemplo:

```python
nums = [10, 20, 30]
nums[1] = 99
```

````python
nums = [10, 20, 30]
print('Antes:', nums)
nums[1] = 99
print('Depois:', nums)
````

## 5. Slicing (fatiamento)

Podemos pegar partes de uma lista com slicing.

Forma geral:

```python
lista[inicio:fim]
```

O índice `inicio` entra, mas o `fim` não entra.

Exemplos:

- `nums[:3]`
- `nums[2:]`
- `nums[1:4]`
- `nums[:]`

````python
nums = [10, 20, 30, 40, 50, 60]
print(nums[:3])
print(nums[2:])
print(nums[1:4])
print(nums[:])
````

Também podemos usar **passo**:

```python
lista[inicio:fim:passo]
```

Exemplos:

- `nums[::2]` → pega de 2 em 2
- `nums[::-1]` → inverte a lista

````python
nums = [1, 2, 3, 4, 5, 6, 7, 8]
print(nums[::2])
print(nums[::-1])
````

## 6. Operadores com listas

### Concatenação com `+`

c = a + b
```

### Repetição com `*`

```

````python
a = [1, 2]
b = [3, 4]
print(a + b)
print([1, 2] * 3)
````

## 7. Funções e métodos importantes

Agora vamos ver funções e métodos muito usados com listas.

### 7.1 `len(lista)`

Retorna a quantidade de elementos.

````python
nums = [10, 20, 30, 40]
print(len(nums))
````

### 7.2 `append(valor)`

Adiciona um elemento ao final da lista.

````python
nums = [10, 20, 30]
nums.append(40)
print(nums)
````

### 7.3 `insert(posicao, valor)`

Insere um valor em uma posição específica.

````python
nums = [10, 20, 30]
nums.insert(1, 15)
print(nums)
````

### 7.4 `remove(valor)`

Remove a primeira ocorrência do valor informado.

````python
nums = [10, 20, 30, 20]
nums.remove(20)
print(nums)
````

### 7.5 `pop()`

Remove o último elemento da lista e o retorna.

````python
nums = [10, 20, 30]
ultimo = nums.pop()
print('Valor removido:', ultimo)
print(nums)
````

Também podemos usar `pop(indice)` para remover um elemento de uma posição específica.

````python
nomes = ['Ana', 'Bruno', 'Carlos']
removido = nomes.pop(1)
print('Removido:', removido)
print(nomes)
````

### 7.6 `sort()`

Ordena a lista.

````python
nums = [5, 2, 9, 1, 7]
nums.sort()
print(nums)
````

### 7.7 `reverse()`

Inverte a ordem dos elementos.

````python
nums = [1, 2, 3, 4]
nums.reverse()
print(nums)
````

### 7.8 `count(valor)`

Conta quantas vezes um valor aparece.

````python
nums = [10, 20, 20, 30, 20]
print(nums.count(20))
````

### 7.9 `index(valor)`

Retorna a posição da primeira ocorrência de um valor.

````python
nomes = ['Ana', 'Bruno', 'Carlos']
print(nomes.index('Bruno'))
````

### 7.10 `in` e `not in`

Podemos verificar se um valor está ou não está na lista.

````python
nums = [1, 2, 3, 4]
print(3 in nums)
print(10 in nums)
print(10 not in nums)
````

## 8. Atenção: métodos que alteram a própria lista

Alguns métodos mudam a lista original, como:

- `append()`
- `insert()`
- `remove()`
- `pop()`
- `sort()`
- `reverse()`

Ou seja, eles **não criam automaticamente uma nova lista**.

## 9. Listas e strings

Listas e strings se relacionam bastante em Python.

### `split()`
Transforma uma string em lista.

### `join()`
Junta uma lista de strings em uma única string.

````python
frase = 'python é muito legal'
palavras = frase.split()
print(palavras)

nova_frase = ' - '.join(palavras)
print(nova_frase)
````
