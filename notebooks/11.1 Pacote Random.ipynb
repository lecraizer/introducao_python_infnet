# Aula: Aleatoriedade em Python (`random`)

Esta aula apresenta o módulo **`random`**, que permite gerar números aleatórios e fazer escolhas aleatórias em Python.

Nela, vamos reutilizar conteúdos que você já viu:

- variáveis
- listas

A ideia é aprender o básico de aleatoriedade e depois praticar com exercícios.

## Importando o módulo

Para usar as funções de aleatoriedade, primeiro importamos o módulo:

```python
import random
```

````python
import random
````

## 1. `random.randint(a, b)`

Gera um número inteiro aleatório entre **`a`** e **`b`**, incluindo os dois extremos.

````python
import random

n = random.randint(1, 10)
print('Número aleatório:', n)
````

## 2. `random.choice(lista)`

Escolhe **um elemento aleatório** de uma lista.

````python
import random

nomes = ['Ana', 'Bruno', 'Carlos', 'Daniel']
sorteado = random.choice(nomes)
print('Nome sorteado:', sorteado)
````

## 3. `random.shuffle(lista)`

Embaralha os elementos de uma lista.

Importante: essa função **altera a própria lista**.

````python
import random

nums = [1, 2, 3, 4, 5]
random.shuffle(nums)
print(nums)
````

## 4. `random.sample(lista, k)`

Seleciona **`k` elementos aleatórios sem repetir**.

````python
import random

nums = [1, 2, 3, 4, 5, 6]
amostra = random.sample(nums, 3)
print(amostra)
````

## Exercício 1 (fácil) — Simular um dado

## Exercício 2 (fácil) — Cara ou coroa

Mostre aleatoriamente `cara` ou `coroa`.
Use uma lista e `random.choice()`.

````python
# Gabarito
import random

moeda = ['cara', 'coroa']
resultado = random.choice(moeda)
print(resultado)
````

## Exercício 3 (fácil) — Sorteio de aluno

Dada uma lista de alunos, sorteie um aluno aleatoriamente.

````python
# Gabarito
import random

alunos = ['Ana', 'Bruno', 'Carlos', 'Daniel', 'Eduardo']
sorteado = random.choice(alunos)
print('Aluno sorteado:', sorteado)
````

## Exercício 4 (médio) — Número secreto

Gere um número aleatório entre **1 e 10**.
Peça para o usuário tentar adivinhar.
Mostre se ele acertou ou não.

````python
# Gabarito
import random

numero = random.randint(1, 10)
palpite = int(input('Adivinhe o número (1 a 10): '))

if palpite == numero:
    print('Acertou!')
else:
    print('Errou! O número era', numero)
````

## Exercício 6 (médio) — Sorteio de 3 números

Dada uma lista de números, sorteie **3 números diferentes**.

````python
# Gabarito
import random

nums = [1, 2, 3, 4, 5, 6, 7, 8, 9]
sorteio = random.sample(nums, 3)
print(sorteio)
````

## Exercício 7 (difícil) — Jogo de adivinhação

O computador escolhe um número entre **1 e 20**.
O usuário tenta adivinhar até acertar.

````python
# Gabarito
import random

numero = random.randint(1, 20)
palpite = 0

while palpite != numero:
    palpite = int(input('Tente adivinhar: '))

print('Acertou!')
````

## Exercício 8 (difícil) — Gerar senha aleatória

Gere uma senha de **6 caracteres aleatórios** usando letras minúsculas e números.

````python
# Gabarito
import random

chars = 'abcdefghijklmnopqrstuvwxyz0123456789'
senha = ''

for i in range(6):
    senha += random.choice(chars)

print('Senha gerada:', senha)
````

## Exercício 9 (difícil) — Loteria

Gere **6 números diferentes** entre 1 e 60.

````python
# Gabarito
import random

nums = list(range(1, 61))
resultado = random.sample(nums, 6)
print(resultado)
````
