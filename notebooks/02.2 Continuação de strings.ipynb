## 1. O que é uma *string*?

Em Python, uma **string** é uma sequência de caracteres: letras, números, espaços, símbolos...

Alguns exemplos de strings:

- `'Python'`
- "Olá, mundo!"
- `'12345'` (mesmo parecendo número, aqui é texto)

Internamente, toda string tem o tipo `str`.


````python
# exemplos de strings
nome = 'Luis'
frase = "Python é legal"
numero_como_texto = '123'

print(nome)
print(frase)
print(numero_como_texto)

print(type(nome))
print(type(frase))
````

## 2.1 Criando strings com aspas simples, duplas e triplas

Você pode criar strings de algumas formas:

- Aspas simples: `'texto aqui'`
- Aspas duplas: `"texto aqui"`
- Aspas triplas: `'''texto aqui'''` ou `"""texto aqui"""` (usadas para textos em várias linhas)

As aspas simples e duplas funcionam quase igual. As triplas permitem **quebrar linha** dentro da string.


````python
# aspas simples e duplas
s1 = 'Olá, mundo!'
s2 = "Python é divertido"

print(s1)
print(s2)

# aspas triplas (string de múltiplas linhas)
texto_longo = '''Este é um texto
que ocupa mais de uma linha.
Legal, né?'''

print(texto_longo)
````

## 2.2 e 2.3 – Juntando (`+`) e repetindo (`*`) strings

- O operador `+` serve para **concatenar** (juntar) strings.
- O operador `*` com um número inteiro serve para **repetir** uma string.

Importante: o `+` só funciona entre strings. Para juntar texto e número, precisamos converter o número para string (veremos isso mais adiante).


````python
# concatenando strings
nome = 'Ana'
sobrenome = 'Silva'

nome_completo = nome + ' ' + sobrenome
print(nome_completo)

# repetindo strings
risada = 'ha'
print(risada * 3)    # 'hahaha'

linha = '-'
print(linha * 30)
````

## 2.4 – Mudando maiúsculas e minúsculas

Strings possuem vários **métodos** (funções especiais ligadas ao objeto). Aqui vamos ver:

- `upper()` – coloca tudo em **MAIÚSCULO**
- `lower()` – coloca tudo em **minúsculo**
- `swapcase()` – inverte maiúsculas/minúsculas
- `capitalize()` – deixa só a **primeira letra** maiúscula
- `title()` – deixa a **primeira letra de cada palavra** maiúscula
- `isupper()` – testa se a string está toda em maiúsculas (retorna `True` ou `False`)
- `islower()` – testa se a string está toda em minúsculas


````python
texto = 'Python é MUITO legal'

print('original:       ', texto)
print('upper():        ', texto.upper())
print('lower():        ', texto.lower())
print('swapcase():     ', texto.swapcase())
print('capitalize():   ', texto.capitalize())
print('title():        ', texto.title())

so_maiusculas = 'ABC'
so_minusculas = 'abc'

print('\nso_maiusculas.isupper():', so_maiusculas.isupper())
print('so_maiusculas.islower():', so_maiusculas.islower())
print('so_minusculas.isupper():', so_minusculas.isupper())
print('so_minusculas.islower():', so_minusculas.islower())
````

## 2.5 – Verificando se um texto contém outro com `in`

O operador `in` verifica se uma **string menor** está contida dentro de outra string.

Ele devolve `True` se encontrar, ou `False` caso contrário.


````python
frase = 'programar em Python é divertido'

print('Python' in frase)
print('Java' in frase)

# também funciona com uma única letra
print('a' in frase)
print('z' in frase)
````

## 2.6 – Contando caracteres e "palavras" com `len()` e `count()`

- A função **`len()`** devolve o **tamanho** da string (quantidade de caracteres, incluindo espaços).
- O método **`count()`** conta quantas vezes um **pedaço de texto** aparece dentro da string.

Para aproximar o número de palavras sem usar outros recursos, podemos contar os espaços e somar 1 (isso funciona bem quando as palavras estão separadas por um único espaço).


````python
frase = 'Python é divertido de aprender'

print('Frase:', frase)
print('Quantidade de caracteres:', len(frase))

# quantas vezes a letra "e" aparece?
print("Quantidade de 'e':", frase.count('e'))

# quantas vezes a palavra 'Python' aparece?
print("Quantidade de 'Python':", frase.count('Python'))

# estimando quantidade de palavras: numero de espaços + 1
qtd_palavras = frase.count(' ') + 1
print('Quantidade aproximada de palavras:', qtd_palavras)
````

## 2.7 e 2.8 – Conversões entre strings, inteiros e floats

Às vezes o valor chega como **texto** (string), mas queremos trabalhar como **número**.

Funções importantes:

- `int(x)` – converte para inteiro (se possível)
- `float(x)` – converte para número com casas decimais
- `str(x)` – converte qualquer coisa para string

**Atenção:** para converter, o texto precisa estar em um formato válido, por exemplo `'42'` ou `'3.14'`.


````python
# string para inteiro
texto_numero = '42'
numero_inteiro = int(texto_numero)
print(texto_numero, '->', numero_inteiro, type(numero_inteiro))

# string para float
texto_pi = '3.1415'
numero_float = float(texto_pi)
print(texto_pi, '->', numero_float, type(numero_float))

# inteiro para string
idade = 30
idade_texto = str(idade)
print(idade, '->', idade_texto, type(idade_texto))

# exemplo de erro comum (descomente para ver o erro):
# numero_errado = int('42a')  # ValueError
````

## 2.9 – Lendo dados do teclado com `input()`

A função **`input()`** lê um texto digitado pelo usuário no terminal e **sempre devolve uma string**.

Forma geral:

```python
nome = input('Digite seu nome: ')
```

Se você precisar de um número, primeiro lê como string e depois converte com `int()` ou `float()`.


````python
# Exemplo 1: lendo um nome
# (roda esta célula e digite algo na caixinha que aparece abaixo no Jupyter)

nome = input('Digite seu nome: ')
print('Olá,', nome + '! Bem-vindo(a) ao Python.')
````

````python
# Exemplo 2: lendo um número inteiro

idade_texto = input('Digite sua idade: ')
idade = int(idade_texto)  # conversão para int

print('Daqui a 5 anos você terá', idade + 5, 'anos.')
````

````python
# Exemplo 3: lendo um número com casas decimais

altura_texto = input('Digite sua altura em metros (ex: 1.75): ')
altura = float(altura_texto)

print('Sua altura é', altura, 'm.')
````
