## Exercício 1 — Par, ímpar ou zero

Peça ao usuário que digite um número inteiro.

Seu programa deve analisar esse número e mostrar uma das seguintes mensagens:

- `Número positivo e par`
- `Número positivo e ímpar`
- `Número negativo`
- `Zero`

### Observações
- Considere que o usuário digitará um valor inteiro válido.
- Use `if`, `elif` e `else` para tratar os diferentes casos.

````python
# Gabarito Exercício 1
n = int(input('Digite um número inteiro: '))

if n > 0 and n % 2 == 0:
    print('Número positivo e par')
elif n > 0 and n % 2 != 0:
    print('Número positivo e ímpar')
elif n < 0:
    print('Número negativo')
else:
    print('Zero')
````

## Exercício 2 — Maior de três números

Peça ao usuário que digite **três números** (um de cada vez).

Depois, determine qual deles é o **maior** e mostre esse valor na tela.

### Exemplo
Se o usuário digitar `5`, `9` e `2`, o programa deve mostrar:

```python
Maior número: 9
```

### Observações
- Considere apenas o caso de números válidos.
- Caso existam empates, basta mostrar o maior valor encontrado.

````python
# Gabarito Exercício 2
n1 = float(input('Digite o primeiro número: '))
n2 = float(input('Digite o segundo número: '))
n3 = float(input('Digite o terceiro número: '))

if n1 >= n2 and n1 >= n3:
    maior = n1
elif n2 >= n1 and n2 >= n3:
    maior = n2
else:
    maior = n3

print('Maior número:', maior)
````

## Exercício 3 — Identificador de letra

Peça ao usuário que digite **um único caractere**.

O programa deve verificar se esse caractere é:

- uma **vogal**
- uma **consoante**
- ou se **não é letra**

### Observações
- Considere vogais: `a`, `e`, `i`, `o`, `u`.
- O programa deve funcionar tanto para letras maiúsculas quanto minúsculas.

## Exercício 5 — Comparação de palavras

Peça ao usuário que digite **duas palavras**.

Depois, analise a relação entre elas e mostre uma mensagem apropriada:

- `Palavras iguais` se forem exatamente iguais
- `Palavras com o mesmo tamanho` se forem diferentes, mas tiverem a mesma quantidade de caracteres
- `Palavras diferentes` nos demais casos

### Observações
- Considere a comparação exatamente como digitada pelo usuário.

## Exercício 6 — Palíndromo simples

Peça ao usuário que digite uma palavra.

Verifique se essa palavra é um **palíndromo**, isto é, se ela pode ser lida da mesma forma da esquerda para a direita e da direita para a esquerda.

### Exemplos
- `arara` → é palíndromo
- `radar` → é palíndromo
- `python` → não é palíndromo

### Observações
- Resolva usando comparação entre a palavra original e sua versão invertida.
- Não é necessário tratar espaços ou acentos.

````python
# Gabarito Exercício 6
palavra = input('Digite uma palavra: ')

if palavra == palavra[::-1]:
    print('É palíndromo')
else:
    print('Não é palíndromo')
````

## Exercício 8 — Horário válido

Peça ao usuário que digite um horário no formato:

```python
hh:mm
```

Verifique se esse horário é aparentemente válido.

Para isso, o programa deve conferir:

- se a string tem tamanho 5
- se existe `:` na posição correta
- se a parte da hora está entre `00` e `23`
- se a parte dos minutos está entre `00` e `59`

Ao final, mostre:

- `Horário válido` ou
- `Horário inválido`

### Observações
- Você pode usar slicing para separar hora e minuto.
- Considere que a entrada pode ter formato incorreto.

````python
# Gabarito Exercício 8
horario = input('Digite um horário no formato hh:mm: ')

if len(horario) == 5 and horario[2] == ':':
    hh = horario[:2]
    mm = horario[3:]

    if hh.isdigit() and mm.isdigit():
        hora = int(hh)
        minuto = int(mm)

        if 0 <= hora <= 23 and 0 <= minuto <= 59:
            print('Horário válido')
        else:
            print('Horário inválido')
    else:
        print('Horário inválido')
else:
    print('Horário inválido')
````

## Exercício 11 — Senha com regra mínima

Peça ao usuário que digite uma senha.

Considere essa senha válida apenas se ela:

- tiver **8 ou mais caracteres**
- contiver **pelo menos um número**

Ao final, mostre:

- `Senha válida` ou
- `Senha inválida`

### Observações
- Como o foco é praticar condicionais, você pode verificar a presença de dígitos usando vários testes com `in`.
- Não é necessário usar laços.

````python
# Gabarito Exercício 11
senha = input('Digite uma senha: ')

tem_numero = (
    '0' in senha or '1' in senha or '2' in senha or '3' in senha or
    '4' in senha or '5' in senha or '6' in senha or '7' in senha or
    '8' in senha or '9' in senha
)

if len(senha) >= 8 and tem_numero:
    print('Senha válida')
else:
    print('Senha inválida')
````

## Exercício 12 — E-mail simples

Peça ao usuário que digite um e-mail.

Considere o e-mail **válido** apenas se:

- contiver `@`
- contiver `.`
- o `@` aparecer antes do `.`
- não começar com `@`
- não terminar com `.`

Ao final, mostre:

- `E-mail válido` ou
- `E-mail inválido`

### Observações
- Esta é uma validação **simples**, não precisa cobrir todos os casos reais de e-mail.
- O objetivo é praticar condições compostas e análise de strings.

````python
# Gabarito Exercício 12
email = input('Digite um e-mail: ')

if (
    '@' in email and
    '.' in email and
    email.index('@') < email.rindex('.') and
    not email.startswith('@') and
    not email.endswith('.')
):
    print('E-mail válido')
else:
    print('E-mail inválido')
````

## Exercício 13 — Data simples

Peça ao usuário que digite uma data no formato:

```python
dd/mm/aaaa
```

Verifique se essa data é aparentemente válida.

Seu programa deve conferir:

- se a string tem tamanho 10
- se há `/` nas posições corretas
- se dia, mês e ano são numéricos
- se o dia está entre `1` e `31`
- se o mês está entre `1` e `12`

Mostre:

- `Data válida` ou
- `Data inválida`

### Observações
- Não é necessário, neste exercício, validar corretamente a quantidade de dias de cada mês.
- Basta a verificação simples acima.

````python
# Gabarito Exercício 13
data = input('Digite uma data no formato dd/mm/aaaa: ')

if len(data) == 10 and data[2] == '/' and data[5] == '/':
    dd = data[:2]
    mm = data[3:5]
    aaaa = data[6:]

    if dd.isdigit() and mm.isdigit() and aaaa.isdigit():
        dia = int(dd)
        mes = int(mm)

        if 1 <= dia <= 31 and 1 <= mes <= 12:
            print('Data válida')
        else:
            print('Data inválida')
    else:
        print('Data inválida')
else:
    print('Data inválida')
````

## Exercício 15 — Valor de compra

Peça ao usuário que digite o valor de uma compra.

Aplique o desconto de acordo com a faixa abaixo:

- até `100` → sem desconto
- acima de `100` até `300` → 10%
- acima de `300` até `500` → 15%
- acima de `500` → 20%

Mostre:

- o valor original
- o percentual de desconto
- o valor final

### Observações
- Use `if`, `elif` e `else` para separar as faixas.

````python
# Gabarito Exercício 15
valor = float(input('Digite o valor da compra: '))

if valor <= 100:
    desconto = 0
elif valor <= 300:
    desconto = 10
elif valor <= 500:
    desconto = 15
else:
    desconto = 20

valor_final = valor * (1 - desconto / 100)

print('Valor original:', valor)
print('Desconto:', str(desconto) + '%')
print('Valor final:', valor_final)
````

## Exercício 16 — Equação do segundo grau

Peça ao usuário os valores de `a`, `b` e `c` de uma equação do segundo grau.

Calcule o discriminante:

```python
delta = b**2 - 4*a*c
```

Depois, informe o tipo de solução da equação:

- `Duas raízes reais` se `delta > 0`
- `Uma raiz real` se `delta == 0`
- `Nenhuma raiz real` se `delta < 0`

### Observações
- Não é necessário calcular as raízes.
- Basta classificar o tipo de resultado a partir do valor de `delta`.

````python
# Gabarito Exercício 16
a = float(input('Digite o valor de a: '))
b = float(input('Digite o valor de b: '))
c = float(input('Digite o valor de c: '))

delta = b**2 - 4*a*c

if delta > 0:
    print('Duas raízes reais')
elif delta == 0:
    print('Uma raiz real')
else:
    print('Nenhuma raiz real')
````

## Exercício 17 — Casa no tabuleiro de xadrez

Peça ao usuário uma posição do tabuleiro de xadrez no formato, por exemplo:

```python
a1
c7
h8
```

Determine se a casa informada é:

- `Casa preta` ou
- `Casa branca`

### Observações
- Considere colunas de `a` até `h` e linhas de `1` até `8`.
- Você pode transformar a letra em número usando comparações simples.
- Uma forma de resolver é somar o índice da coluna com o número da linha e analisar se o resultado é par ou ímpar.

````python
# Gabarito Exercício 17
posicao = input('Digite uma posição do tabuleiro (ex: a1): ').lower()

coluna = posicao[0]
linha = int(posicao[1])

if coluna == 'a':
    c = 1
elif coluna == 'b':
    c = 2
elif coluna == 'c':
    c = 3
elif coluna == 'd':
    c = 4
elif coluna == 'e':
    c = 5
elif coluna == 'f':
    c = 6
elif coluna == 'g':
    c = 7
else:
    c = 8

if (c + linha) % 2 == 0:
    print('Casa preta')
else:
    print('Casa branca')
````

## Exercício 19 — Resultado de partida

Peça ao usuário:

- o nome do time 1
- o número de gols do time 1
- o nome do time 2
- o número de gols do time 2

Depois, mostre o placar final e classifique o resultado da partida como:

- `Jogo empatado` se os dois times fizerem a mesma quantidade de gols
- `Jogo equilibrado` se a vitória acontecer por apenas 1 gol de diferença
- `Vitória convincente` se a diferença for de 2 ou mais gols

Além disso, quando houver vencedor, mostre também o nome do time vencedor.

### Exemplo
```python
Flamengo 2 x 1 Vasco
Vencedor: Flamengo
Jogo equilibrado
```

````python
# Gabarito Exercício 19
time1 = input('Digite o nome do time 1: ')
gols1 = int(input('Digite os gols do time 1: '))
time2 = input('Digite o nome do time 2: ')
gols2 = int(input('Digite os gols do time 2: '))

print(time1, gols1, 'x', gols2, time2)

if gols1 == gols2:
    print('Jogo empatado')
elif gols1 > gols2:
    print('Vencedor:', time1)
    if gols1 - gols2 == 1:
        print('Jogo equilibrado')
    else:
        print('Vitória convincente')
else:
    print('Vencedor:', time2)
    if gols2 - gols1 == 1:
        print('Jogo equilibrado')
    else:
        print('Vitória convincente')
````

## Exercício 20 — Semáforo inteligente

Peça ao usuário duas informações:

- a cor do semáforo (`verde`, `amarelo` ou `vermelho`)
- se há pedestre atravessando (`sim` ou `nao`)

Com base nessas informações, mostre a ação recomendada:

- `verde` + `nao` → `Pode seguir`
- `verde` + `sim` → `Reduza e aguarde`
- `amarelo` + `nao` → `Atenção`
- `amarelo` + `sim` → `Pare com segurança`
- `vermelho` em qualquer caso → `Pare`

### Observações
- Converta as entradas para minúsculas, se quiser, para facilitar a comparação.

````python
# Gabarito Exercício 20
cor = input('Digite a cor do semáforo: ').lower()
pedestre = input('Há pedestre atravessando? (sim/nao): ').lower()

if cor == 'verde' and pedestre == 'nao':
    print('Pode seguir')
elif cor == 'verde' and pedestre == 'sim':
    print('Reduza e aguarde')
elif cor == 'amarelo' and pedestre == 'nao':
    print('Atenção')
elif cor == 'amarelo' and pedestre == 'sim':
    print('Pare com segurança')
else:
    print('Pare')
````
