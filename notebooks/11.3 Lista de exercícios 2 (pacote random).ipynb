## Import inicial

Antes de começar, importe o módulo `random`.


```python
import random
```

## Exercício 1 — Simulação de moedas

Simule **100 lançamentos de moeda**.

Use:

- `0 = cara`
- `1 = coroa`

O programa deve contar:

- quantas vezes saiu **cara**
- quantas vezes saiu **coroa**

No final, mostrar os resultados.


```python
cara = 0
coroa = 0

for i in range(100):
    moeda = random.randint(0, 1)

    if moeda == 0:
        cara += 1
    else:
        coroa += 1

print("Quantidade de caras:", cara)
print("Quantidade de coroas:", coroa)
```

## Exercício 2 — Maior e menor número sorteado

O programa deve gerar **20 números aleatórios entre 1 e 100**.

Durante os sorteios, o programa deve:

- mostrar os números gerados
- guardar todos em uma lista

No final, mostrar:

- o **maior número** sorteado
- o **menor número** sorteado


```python
numeros = []

for i in range(20):
    numero = random.randint(1, 100)
    numeros.append(numero)
    print(numero)

maior = numeros[0]
menor = numeros[0]

for numero in numeros:
    if numero > maior:
        maior = numero
    if numero < menor:
        menor = numero

print("\nLista completa:", numeros)
print("Maior número:", maior)
print("Menor número:", menor)
```

## Exercício 3 — Soma até ultrapassar 50

O programa deve gerar números aleatórios entre **1 e 10**.

Os números devem ser somados até que a soma total **ultrapasse 50**.

O programa deve:

- mostrar cada número sorteado
- mostrar a soma acumulada
- parar quando a soma passar de 50

No final, mostrar:

- a soma final
- quantos números foram gerados


```python
soma = 0
quantidade = 0

while soma <= 50:
    numero = random.randint(1, 10)
    soma += numero
    quantidade += 1

    print("Número sorteado:", numero)
    print("Soma acumulada:", soma)
    print("---")

print("Soma final:", soma)
print("Quantidade de números gerados:", quantidade)
```

## Exercício 4 — Frequência dos números

Gere **30 números aleatórios entre 1 e 5**.

O programa deve contar **quantas vezes cada número apareceu**.


```python
cont_1 = 0
cont_2 = 0
cont_3 = 0
cont_4 = 0
cont_5 = 0

for i in range(30):
    numero = random.randint(1, 5)
    print(numero, end=" ")

    if numero == 1:
        cont_1 += 1
    elif numero == 2:
        cont_2 += 1
    elif numero == 3:
        cont_3 += 1
    elif numero == 4:
        cont_4 += 1
    else:
        cont_5 += 1

print("\n")
print("Número 1 apareceu", cont_1, "vezes")
print("Número 2 apareceu", cont_2, "vezes")
print("Número 3 apareceu", cont_3, "vezes")
print("Número 4 apareceu", cont_4, "vezes")
print("Número 5 apareceu", cont_5, "vezes")
```

## Exercício 5 — Letras aleatórias

O programa deve gerar **10 letras aleatórias do alfabeto**.

As letras devem ser armazenadas em uma lista.

No final, o programa deve mostrar:

- todas as letras geradas
- quantas são **vogais**
- quantas são **consoantes**


```python
alfabeto = "abcdefghijklmnopqrstuvwxyz"
letras = []
vogais = 0
consoantes = 0

for i in range(10):
    letra = random.choice(alfabeto)
    letras.append(letra)

    if letra in "aeiou":
        vogais += 1
    else:
        consoantes += 1

print("Letras geradas:", letras)
print("Quantidade de vogais:", vogais)
print("Quantidade de consoantes:", consoantes)
```

## Exercício 6 — Lista de números únicos

O programa deve gerar números aleatórios entre **1 e 20**, mas **não pode repetir números**.

Os números devem ser guardados em uma **lista**.

O programa deve continuar gerando números até que a lista tenha **10 números diferentes**.

No final, mostrar a lista completa.


```python
numeros = []

while len(numeros) < 10:
    numero = random.randint(1, 20)

    if numero not in numeros:
        numeros.append(numero)

print("Lista final:", numeros)
```

## Exercício 7 — Até aparecer três vezes seguidas

O programa deve gerar números aleatórios entre **1 e 6** (como um dado).

O sorteio deve continuar até que **o mesmo número apareça três vezes seguidas**.

O programa deve:

- mostrar todos os números gerados
- parar quando três números iguais consecutivos aparecerem
- mostrar quantos sorteios foram necessários


```python
numeros = []
quantidade = 0

while True:
    numero = random.randint(1, 6)
    numeros.append(numero)
    quantidade += 1
    print(numero)

    if len(numeros) >= 3:
        if numeros[-1] == numeros[-2] == numeros[-3]:
            break

print("\nTrês números iguais consecutivos encontrados!")
print("Total de sorteios:", quantidade)
```

## Exercício 8 — Sequência crescente

Gere números aleatórios entre **1 e 50** até aparecer uma **sequência crescente de 3 números seguidos**.

Exemplo que faria o programa parar:

`12, 8, 14, 19, 22`

Nesse caso, `14 → 19 → 22` é crescente.

Quando isso acontecer, o programa deve:

- parar
- informar quantos sorteios foram feitos


```python
numeros = []
quantidade = 0

while True:
    numero = random.randint(1, 50)
    numeros.append(numero)
    quantidade += 1
    print(numero)

    if len(numeros) >= 3:
        if numeros[-3] < numeros[-2] and numeros[-2] < numeros[-1]:
            break

print("\nSequência crescente de 3 números encontrada!")
print("Total de sorteios:", quantidade)
```

## Exercício 9 — Número mais repetido

O programa deve gerar **40 números aleatórios entre 1 e 10** e guardar em uma lista.

Depois disso, o programa deve descobrir **qual número apareceu mais vezes**.


```python
numeros = []

for i in range(40):
    numero = random.randint(1, 10)
    numeros.append(numero)

mais_frequente = numeros[0]
maior_quantidade = 0

for candidato in range(1, 11):
    contagem = 0

    for numero in numeros:
        if numero == candidato:
            contagem += 1

    if contagem > maior_quantidade:
        maior_quantidade = contagem
        mais_frequente = candidato

print("Lista gerada:")
print(numeros)
print("\nNúmero mais frequente:", mais_frequente)
print("Quantidade de vezes:", maior_quantidade)
```

## Exercício 10 — Sequência específica

O programa deve gerar números aleatórios entre **1 e 5** até aparecer a sequência:

`3, 1, 4`

Ou seja, três sorteios consecutivos exatamente nessa ordem.

Quando a sequência aparecer, o programa deve:

- parar
- mostrar quantos sorteios foram feitos


```python
numeros = []
quantidade = 0

while True:
    numero = random.randint(1, 5)
    numeros.append(numero)
    quantidade += 1
    print(numero)

    if len(numeros) >= 3:
        if numeros[-3] == 3 and numeros[-2] == 1 and numeros[-1] == 4:
            break

print("\nA sequência 3, 1, 4 apareceu!")
print("Total de sorteios:", quantidade)
```
