## Exercício 11 — Remover números menores que um limite usando while

```python
lista = [10, 2, 30, 4, 50]
limite = 10
i = 0
while i < len(lista):
    if lista[i] < limite:
        lista.pop(i)
    else:
        i += 1
lista
```

## Exercício 12 — Inverter cada palavra da frase

```python
frase = "python é divertido"
palavras = frase.split()
invertidas = []
for p in palavras:
    invertidas.append(p[::-1])
" ".join(invertidas)
```

## Exercício 13 — Intercalar duas listas

```python
l1 = ["Ana", "Bruno", "Clara"]
l2 = [20, 25, 30]
res = []
for i in range(len(l1)):
    res.append(l1[i])
    res.append(l2[i])
res
```

## Exercício 14 — Gerar combinações de palavras

```python
l1 = ["doce", "salgado"]
l2 = ["de festa", "simples"]
res = []
for a in l1:
    for b in l2:
        res.append(a + " " + b)
res
```

## Exercício 15 — Compressão RLE

```python
s = "aaabbcaaa"
res = ""
i = 0
while i < len(s):
    letra = s[i]
    count = 1
    j = i + 1
    while j < len(s) and s[j] == letra:
        count += 1
        j += 1
    res += letra + str(count)
    i = j
res
```

## Exercício 16 — Mini Jogo da Forca (versão simples)

```python
palavra = "python"
mostrar = ["_"] * len(palavra)
tentadas = []
erros = 0
limite = 6

while erros < limite and "_" in mostrar:
    print("Letras tentadas:", tentadas)
    print("Palavra:", " ".join(mostrar))
    letra = input("Digite uma letra: ")
    
    if letra == '0':
        print('Você escolheu sair do programa.')
        break

    if len(letra) > 1:
        print('Escolha apenas uma letra.')
        continue

    if not letra.isalpha():
        print('Digite uma letra.')
        continue
    
    if letra in tentadas:
        print("Você já tentou essa letra.")
        continue

    tentadas.append(letra)

    if letra in palavra:
        for i in range(len(palavra)):
            if palavra[i] == letra:
                mostrar[i] = letra
    else:
        erros += 1
        print("Letra incorreta.")

    
if "_" not in mostrar:
    print("Você venceu! A palavra era:", palavra)
else:
    print("Você perdeu! A palavra era:", palavra)
```
