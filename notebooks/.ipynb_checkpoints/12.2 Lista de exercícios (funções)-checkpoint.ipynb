## Exercício 1 — Saudação customizável

### Enunciado

Crie uma função `saudacao(nome, formal=False)`.

Regras:
- se `formal=True`, a função deve retornar `"Olá, Sr(a). <nome>"`
- caso contrário, deve retornar `"Oi, <nome>"`

### Gabarito

```python
# Gabarito — Exercício 1

def saudacao(nome, formal=False):
    if formal:
        return f"Olá, Sr(a). {nome}"
    return f"Oi, {nome}"

print(saudacao("Luis"))
print(saudacao("Luis", True))
```

## Exercício 2 — Soma com valor mínimo

### Enunciado

Crie uma função que recebe:
- uma lista de números
- um valor mínimo

A função deve retornar a soma apenas dos valores da lista que sejam **maiores ou iguais** ao valor mínimo.

### Gabarito

```python
# Gabarito — Exercício 2

def soma_com_minimo(lista, minimo):
    soma = 0
    for numero in lista:
        if numero >= minimo:
            soma += numero
    return soma

print(soma_com_minimo([2, 5, 8, 1, 10], 5))
```

## Exercício 3 — Inverter palavras

### Enunciado

Crie uma função que recebe uma frase e retorna uma nova frase em que:
- cada palavra foi invertida individualmente
- a ordem das palavras foi mantida

Exemplo:
`"python é legal"` → `"nohtyp é lagel"`

### Gabarito

```python
# Gabarito — Exercício 3

def inverter_palavras(frase):
    palavras = frase.split()
    palavras_invertidas = []

    for palavra in palavras:
        palavras_invertidas.append(palavra[::-1])

    return " ".join(palavras_invertidas)

print(inverter_palavras("python é legal"))
```

## Exercício 4 — Validação de senha

### Enunciado

Crie uma função que recebe uma senha e retorna `True` ou `False`.

Regras:
- deve ter pelo menos 8 caracteres
- deve ter pelo menos 1 número
- deve ter pelo menos 1 letra

### Gabarito

```python
# Gabarito — Exercício 4

def validar_senha(senha):
    if len(senha) < 8:
        return False

    tem_numero = False
    tem_letra = False

    for caractere in senha:
        if caractere.isdigit():
            tem_numero = True
        if caractere.isalpha():
            tem_letra = True

    return tem_numero and tem_letra

print(validar_senha("abc123"))
print(validar_senha("abc12345"))
```

## Exercício 5 — Jogo de adivinhação modularizado

### Enunciado

Crie:
1. uma função que gera um número aleatório entre 1 e 100
2. uma função que recebe um palpite e o número secreto e retorna:
   - `"maior"` se o número secreto for maior
   - `"menor"` se o número secreto for menor
   - `"acertou"` se forem iguais

Depois, use essas funções para montar um pequeno jogo.

### Gabarito

```python
# Gabarito — Exercício 5

import random

def gerar_numero():
    return random.randint(1, 100)

def verificar_palpite(palpite, secreto):
    if palpite < secreto:
        return "maior"
    elif palpite > secreto:
        return "menor"
    return "acertou"

def jogar():
    secreto = gerar_numero()

    while True:
        palpite = int(input("Digite um número entre 1 e 100: "))
        resultado = verificar_palpite(palpite, secreto)
        print(resultado)

        if resultado == "acertou":
            break

# Para jogar, descomente:
# jogar()
```

## Exercício 6 — Média com pesos opcionais

### Enunciado

Crie uma função que recebe:
- uma lista de números
- uma lista de pesos (opcional)

Regras:
- se não receber pesos, calcule a média simples
- se receber pesos, calcule a média ponderada

### Gabarito

```python
# Gabarito — Exercício 6

def media(lista, pesos=None):
    if pesos is None:
        return sum(lista) / len(lista)

    soma_ponderada = 0
    soma_pesos = 0

    for i in range(len(lista)):
        soma_ponderada += lista[i] * pesos[i]
        soma_pesos += pesos[i]

    return soma_ponderada / soma_pesos

print(media([7, 8, 9]))
print(media([7, 8, 9], [1, 2, 3]))
```

## Exercício 7 — Contar vogais

### Enunciado

Crie uma função que recebe uma string e retorna quantas vogais ela possui.

Considere:
- vogais minúsculas e maiúsculas
- vogais com acento podem ser ignoradas

### Gabarito

```python
# Gabarito — Exercício 7

def contar_vogais(texto):
    total = 0
    vogais = "aeiouAEIOU"

    for caractere in texto:
        if caractere in vogais:
            total += 1

    return total

print(contar_vogais("Python é muito legal"))
```

## Exercício 8 — Remover repetidos mantendo a ordem

### Enunciado

Crie uma função que recebe uma lista e retorna uma nova lista sem elementos repetidos,
mas mantendo a ordem da primeira vez em que cada elemento apareceu.

Exemplo:
`[4, 2, 4, 1, 2, 7, 1]` → `[4, 2, 1, 7]`

### Gabarito

```python
# Gabarito — Exercício 8

def remover_repetidos(lista):
    nova_lista = []

    for elemento in lista:
        if elemento not in nova_lista:
            nova_lista.append(elemento)

    return nova_lista

print(remover_repetidos([4, 2, 4, 1, 2, 7, 1]))
```

## Exercício 9 — Maior sequência de mesmo caractere

### Enunciado

Crie uma função que recebe uma string e retorna o tamanho da maior sequência consecutiva
do mesmo caractere.

Exemplos:
- `"aaabbccccd"` → `4`
- `"abc"` → `1`
- `"aabbbccaaaa"` → `4`

### Gabarito

```python
# Gabarito — Exercício 9

def maior_sequencia(texto):
    if texto == "":
        return 0

    maior = 1
    atual = 1

    for i in range(1, len(texto)):
        if texto[i] == texto[i - 1]:
            atual += 1
            if atual > maior:
                maior = atual
        else:
            atual = 1

    return maior

print(maior_sequencia("aaabbccccd"))
print(maior_sequencia("abc"))
print(maior_sequencia("aabbbccaaaa"))
```

## Exercício 10 — Simulação de batalha por rodadas

### Enunciado

Crie uma função que simule uma batalha entre dois jogadores.

A função deve receber:
- vida inicial do jogador 1
- vida inicial do jogador 2
- dano mínimo por rodada
- dano máximo por rodada

Em cada rodada:
- cada jogador causa um dano aleatório no outro
- o dano deve estar entre o mínimo e o máximo

A função deve retornar:
- quem venceu (`"Jogador 1"` ou `"Jogador 2"`)
- e em quantas rodadas a batalha terminou

Você pode supor que os dois atacam em toda rodada.

### Gabarito

```python
# Gabarito — Exercício 10

import random

def batalha(vida1, vida2, dano_min, dano_max):
    rodadas = 0

    while vida1 > 0 and vida2 > 0:
        rodadas += 1

        dano1 = random.randint(dano_min, dano_max)
        dano2 = random.randint(dano_min, dano_max)

        vida2 -= dano1
        vida1 -= dano2

    if vida1 > 0:
        vencedor = "Jogador 1"
    else:
        vencedor = "Jogador 2"

    return vencedor, rodadas

print(batalha(30, 30, 3, 8))
```
