## Exercício 1 — Contar vogais em uma frase

Dada uma string, conte quantas vogais minúsculas (`a`, `e`, `i`, `o`, `u`) existem no texto.

Exemplo:  
Entrada: `"python é muito legal"` → Saída esperada: `9`

## Exercício 2 — Contar quantas vezes uma palavra aparece

Dada uma frase e uma palavra, conte quantas vezes a palavra aparece na frase.

## Exercício 3 — Verificar se uma palavra está na frase

Leia uma frase e uma palavra e diga se ela aparece na frase.  
Mostre apenas o valor lógico (`True` ou `False`).

## Exercício 4 — Substituir todas as ocorrências de uma palavra

Dada uma frase, troque uma palavra por outra.  
Exemplo: trocar `"bolo"` por `"torta"` na frase `"eu gosto de bolo de chocolate"`.

## Exercício 5 — Contar quantos espaços existem em uma frase

Dada uma frase, conte quantos espaços em branco existem.

## Exercício 6 — Deixar só a primeira letra de cada palavra maiúscula

Dada uma frase em letras minúsculas, transforme-a para o formato “Título”, com a primeira letra de cada palavra em maiúscula.

Exemplo: `"python e dados"` → `"Python E Dados"`

## Exercício 7 — Remover todos os espaços da frase

Dada uma frase, crie uma nova string sem nenhum espaço.

Exemplo: `"a b c d"` → `"abcd"`

## Exercício 8 — Quantidade de caracteres sem espaços

Dada uma frase, calcule quantos caracteres **não são espaços**.  
Use a ideia: tamanho total − quantidade de espaços.

## Exercício 9 — Letras 'a' e proporção no texto

Dada uma string, conte quantas letras `"a"` minúsculas existem e calcule a proporção em relação ao tamanho total do texto.

## Exercício 10 — Formatar número de telefone

Dada uma string com 11 dígitos representando um celular no formato `DDNNNNNNNNN`, monte a versão formatada:

Exemplo:  
Entrada: `"21987654321"` → Saída: `"(21) 98765-4321"`

## Exercício 11 — Formatar uma data (DDMMAAAA → DD/MM/AAAA)

Dada uma string contendo uma data somente em números no formato `"DDMMAAAA"`, transforme-a em `"DD/MM/AAAA"`.

## Exercício 12 — Extrair usuário e domínio de um e‑mail

Dado um e-mail no formato `"usuario@dominio.com"`, separe a parte antes e depois do `"@"`, exibindo:

`Usuário: ...`  
`Domínio: ...`

## Exercício 13 — Remover todas as vogais do texto

Peça ao usuário um texto e remova todas as vogais minúsculas (`a`, `e`, `i`, `o`, `u`).  
A nova string deve manter todos os outros caracteres.

## Exercício 14 — Criar um “bordão” com a primeira e última letra

Peça ao usuário uma palavra e construa uma nova string assim:

primeira letra + (várias `*`) + última letra

A quantidade de `*` deve ser igual a (tamanho da palavra − 2).

## Exercício 15 — Frase abreviada com a palavra do meio

Peça ao usuário uma frase com **3 palavras** e gere a forma abreviada:

`primeira (x.) terceira`

Onde `x` é a inicial da palavra do meio.

Use apenas `find()` e fatiamento (`[:]`).
