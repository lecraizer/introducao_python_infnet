## 1. O que é uma função?

Uma **função** é um bloco de código criado para realizar uma tarefa específica.

Em vez de escrever a mesma lógica várias vezes, podemos colocá-la dentro de uma função e chamá-la sempre que precisarmos.

### Ideia intuitiva
Pense em função como uma “máquina”:

- ela pode receber valores de entrada;
- faz algum processamento;
- e pode devolver um resultado.

### Vantagens de usar funções

As funções ajudam a:

- **evitar repetição de código**;
- **organizar melhor** o programa;
- **dividir problemas grandes em partes menores**;
- facilitar correções e testes;
- deixar o código mais legível.

## 2. Estrutura básica de uma função

A forma mais simples de criar uma função é usando `def`.

```python
def nome_da_funcao():
    # código da função
```

Depois de definir a função, podemos chamá-la com:

```python
nome_da_funcao()
```

````python
def dizer_oi():
    print("Oi!")

dizer_oi()
````

No exemplo acima:

- `def` inicia a definição da função;
- `dizer_oi` é o nome da função;
- os parênteses `()` aparecem tanto na definição quanto na chamada;
- o bloco identado pertence à função.

## 3. Primeiro contato: função sem parâmetros

Uma função sem parâmetros é aquela que sempre faz a mesma coisa, sem receber valores externos.

````python
def mostrar_linha():
    print("-" * 30)

mostrar_linha()
mostrar_linha()
````

Esse é um caso simples, mas já mostra um ponto importante:

sem função, talvez tivéssemos que escrever `print("-" * 30)` várias vezes.
Com função, basta chamar `mostrar_linha()`.

````python
def mensagem_boas_vindas():
    print("Bem-vindo ao estudo de funções!")

mensagem_boas_vindas()
````

## 4. Funções com parâmetros

Na prática, muitas funções precisam trabalhar com valores diferentes a cada chamada.

Para isso usamos **parâmetros**.

````python
def saudacao(nome):
    print("Olá,", nome)

saudacao("Luis")
saudacao("Maria")
saudacao("João")
````

Aqui:

- `nome` é o **parâmetro** da função;
- `"Luis"`, `"Maria"` e `"João"` são **argumentos** passados na chamada.

### Parâmetro x argumento

Essa distinção é útil:

- **parâmetro**: variável escrita na definição da função;
- **argumento**: valor real enviado para a função.

````python
def mostrar_soma(a, b):
    print(a + b)

mostrar_soma(3, 5)
mostrar_soma(10, 7)
````

## 5. Mais de um parâmetro

Uma função pode receber vários valores.

````python
def apresentar_pessoa(nome, idade, cidade):
    print("Nome:", nome)
    print("Idade:", idade)
    print("Cidade:", cidade)

apresentar_pessoa("Luis", 30, "Rio de Janeiro")
````

Perceba que a **ordem** importa.

Se trocarmos a ordem dos argumentos, mudamos o significado da chamada.

````python
def apresentar_pessoa(nome, idade, cidade):
    print("Nome:", nome)
    print("Idade:", idade)
    print("Cidade:", cidade)

apresentar_pessoa("Rio de Janeiro", "Luis", 30)
````

O código acima até roda, mas o resultado sai “embaralhado”, porque os argumentos foram passados em ordem errada.

## 6. A importância do `return`

Esse é um dos pontos mais importantes do tema.

Muitos iniciantes confundem `print()` com `return`, mas eles têm papéis diferentes.

### `print()`
Serve para **mostrar** algo na tela.

### `return`
Serve para **devolver** um valor para quem chamou a função.

````python
def soma_print(a, b):
    print(a + b)

resultado = soma_print(2, 3)
print("Valor guardado em resultado:", resultado)
````

Mesmo tendo aparecido `5` na tela, a variável `resultado` recebeu `None`.

Isso aconteceu porque a função apenas imprimiu, mas **não retornou** o valor.

````python
def soma_return(a, b):
    return a + b

resultado = soma_return(2, 3)
print("Valor guardado em resultado:", resultado)
````

Agora a função realmente devolve o valor.

### Regra prática
Se você quer **usar o resultado depois**, normalmente precisa de `return`.

````python
def quadrado(x):
    return x ** 2

print(quadrado(4))
print(quadrado(3) + quadrado(5))
````

Perceba que o retorno pode ser usado:

- em uma variável;
- em outra conta;
- dentro de outra função;
- em `if`, `while`, etc.

## 7. `print` x `return` lado a lado

````python
def dobro_com_print(x):
    print(x * 2)

def dobro_com_return(x):
    return x * 2

a = dobro_com_print(4)
b = dobro_com_return(4)

print("a =", a)
print("b =", b)
````

O `print` apenas exibe.

O `return` devolve o valor para o programa.

## 8. Funções que retornam texto, número ou booleano

Uma função pode retornar vários tipos de valor.

````python
def mensagem(nome):
    return "Olá, " + nome

def triplo(x):
    return x * 3

def eh_par(x):
    return x % 2 == 0

print(mensagem("Luis"))
print(triplo(7))
print(eh_par(10))
print(eh_par(9))
````

## 9. Funções com `if` e `else`

Funções combinam muito bem com estruturas condicionais.

````python
def maior_de_idade(idade):
    if idade >= 18:
        return True
    else:
        return False

print(maior_de_idade(20))
print(maior_de_idade(15))
````

````python
def situacao_nota(nota):
    if nota >= 7:
        return "Aprovado"
    else:
        return "Reprovado"

print(situacao_nota(8.5))
print(situacao_nota(5.0))
````

Muitas vezes a função serve justamente para “encapsular” uma regra.

## 10. Funções com `for`

Também é muito comum criar funções que percorrem listas ou strings.

````python
def somar_lista(lista):
    soma = 0
    for numero in lista:
        soma += numero
    return soma

print(somar_lista([1, 2, 3, 4, 5]))
````

````python
def contar_letras_a(texto):
    contador = 0
    for caractere in texto:
        if caractere == "a":
            contador += 1
    return contador

print(contar_letras_a("banana"))
````

## 11. Funções com `while`

Também podemos usar `while` normalmente dentro de funções.

````python
def contar_ate_n(n):
    i = 1
    while i <= n:
        print(i)
        i += 1

contar_ate_n(5)
````

````python
def soma_1_ate_n(n):
    i = 1
    soma = 0
    while i <= n:
        soma += i
        i += 1
    return soma

print(soma_1_ate_n(5))
````

## 12. Funções que trabalham com strings

Strings aparecem muito em exercícios de funções.

````python
def primeira_letra(texto):
    return texto[0]

print(primeira_letra("Python"))
````

````python
def gritar(texto):
    return texto.upper() + "!!!"

print(gritar("oi"))
````

````python
def inverter_texto(texto):
    return texto[::-1]

print(inverter_texto("python"))
````

## 13. Funções que trabalham com listas

Listas também são extremamente comuns.

````python
def ultimo_elemento(lista):
    return lista[-1]

print(ultimo_elemento([10, 20, 30, 40]))
````

````python
def maior_valor(lista):
    maior = lista[0]

    for numero in lista:
        if numero > maior:
            maior = numero

    return maior

print(maior_valor([4, 9, 2, 15, 7]))
````

````python
def contar_pares(lista):
    total = 0

    for numero in lista:
        if numero % 2 == 0:
            total += 1

    return total

print(contar_pares([1, 2, 3, 4, 6, 7, 9]))
````

## 14. Parâmetros opcionais

Agora entra um ponto bem importante: **argumentos opcionais**.

Podemos definir um valor padrão para um parâmetro.

Assim, se a pessoa não passar esse argumento, a função usa o valor padrão.

````python
def saudacao(nome, formal=False):
    if formal:
        return f"Olá, Sr(a). {nome}"
    else:
        return f"Oi, {nome}"

print(saudacao("Luis"))
print(saudacao("Luis", True))
````

Nesse exemplo:

- `formal=False` significa que, se nada for passado, o valor de `formal` será `False`.

````python
def potencia(base, expoente=2):
    return base ** expoente

print(potencia(5))
print(potencia(5, 3))
````

Aqui:

- `potencia(5)` usa o padrão `expoente=2`;
- `potencia(5, 3)` substitui o valor padrão.

## 15. Regras dos parâmetros opcionais

Em Python, parâmetros opcionais devem vir **depois** dos obrigatórios.

Exemplo correto:

````python
def exemplo_correto(nome, idade=18):
    return f"{nome} tem {idade} anos"

print(exemplo_correto("Ana"))
print(exemplo_correto("Carlos", 25))
````

Exemplo incorreto:

```python
def exemplo_errado(idade=18, nome):
    ...
```

Isso gera erro de sintaxe.

## 16. Chamando funções com argumentos nomeados

Além da ordem normal, podemos passar argumentos pelo nome.

````python
def apresentar(nome, idade, cidade):
    return f"{nome} tem {idade} anos e mora em {cidade}"

print(apresentar("Luis", 30, "Rio"))
print(apresentar(cidade="Rio", nome="Luis", idade=30))
````

Isso pode deixar a chamada mais clara em alguns casos.

## 17. Uma função pode chamar outra

Esse é um ponto muito importante para organizar programas maiores.

Podemos dividir um problema em pequenas funções.

````python
def quadrado(x):
    return x ** 2

def soma_dos_quadrados(a, b):
    return quadrado(a) + quadrado(b)

print(soma_dos_quadrados(3, 4))
````

Essa ideia de quebrar o problema em partes menores é central no uso de funções.

````python
def eh_par(x):
    return x % 2 == 0

def contar_pares(lista):
    total = 0
    for numero in lista:
        if eh_par(numero):
            total += 1
    return total

print(contar_pares([1, 2, 3, 4, 5, 6]))
````

## 18. Escopo de variáveis

Nem toda variável existe em qualquer lugar do programa.

### Variável local
É criada dentro da função e só existe ali.

````python
def teste():
    x = 10
    print("Dentro da função:", x)

teste()
````

Se tentarmos acessar `x` fora da função, teremos erro, porque ela é local.

````python
def teste():
    x = 10
    print("Dentro da função:", x)

teste()

# Descomente a linha abaixo para ver o erro:
# print(x)
````

## 19. Variável de fora da função

Uma função pode ler variáveis externas em muitos casos, mas é melhor ter cuidado.

Do ponto de vista didático, o ideal no começo é:

> preferir passar valores por parâmetro, em vez de depender de variáveis externas.

````python
mensagem = "Olá"

def mostrar():
    print(mensagem)

mostrar()
````

Isso funciona, mas geralmente é melhor algo como:

````python
def mostrar(mensagem):
    print(mensagem)

mostrar("Olá")
````

Assim a função fica mais clara, mais reutilizável e menos dependente do resto do código.

## 20. `return` encerra a função

Quando a execução encontra um `return`, a função termina naquele ponto.

````python
def exemplo_return(x):
    if x > 0:
        return "positivo"
    return "zero ou negativo"

print(exemplo_return(5))
print(exemplo_return(0))
print(exemplo_return(-2))
````

````python
def verificar_numero(x):
    if x % 2 == 0:
        return "par"
    print("Esta linha só roda se o número for ímpar")
    return "ímpar"

print(verificar_numero(8))
print(verificar_numero(7))
````

## 21. Retornando mais de um valor

Em Python, uma função pode retornar vários valores de uma vez.

````python
def dividir_em_duas_partes(n):
    metade = n / 2
    dobro = n * 2
    return metade, dobro

a, b = dividir_em_duas_partes(10)
print(a)
print(b)
````

Tecnicamente, Python está retornando uma tupla, mas neste momento o mais importante é perceber que dá para “desempacotar” o retorno.

## 22. Exemplos úteis e bem didáticos

### Exemplo 1 — contar quantos números são maiores que um valor

````python
def contar_maiores(lista, limite):
    contador = 0

    for numero in lista:
        if numero > limite:
            contador += 1

    return contador

print(contar_maiores([2, 8, 5, 10, 1, 7], 5))
````

### Exemplo 2 — remover espaços do começo e do fim

````python
def limpar_texto(texto):
    return texto.strip()

print(limpar_texto("   oi   "))
````

### Exemplo 3 — verificar se uma palavra é palíndromo

````python
def eh_palindromo(texto):
    texto = texto.lower().replace(" ", "")
    return texto == texto[::-1]

print(eh_palindromo("arara"))
print(eh_palindromo("Ame a ema"))
print(eh_palindromo("python"))
````

### Exemplo 4 — média simples

````python
def media(lista):
    return sum(lista) / len(lista)

print(media([7, 8, 9]))
````

### Exemplo 5 — média com parâmetro opcional

````python
def media_personalizada(lista, arredondar=False):
    m = sum(lista) / len(lista)

    if arredondar:
        return round(m, 2)

    return m

print(media_personalizada([7, 8, 9]))
print(media_personalizada([7, 8, 9], True))
````

## 23. Funções e `random`

Como vocês já viram `random`, vale conectar isso com funções também.

````python
import random

def lançar_dado():
    return random.randint(1, 6)

print(lançar_dado())
print(lançar_dado())
print(lançar_dado())
````

````python
import random

def sortear_nome(lista):
    return random.choice(lista)

nomes = ["Ana", "Bruno", "Carlos", "Duda"]
print(sortear_nome(nomes))
````

## 24. Funções e modularização

Modularizar significa dividir o programa em partes menores.

Exemplo: em vez de fazer um jogo inteiro em um único bloco enorme, podemos criar várias funções pequenas.

````python
import random

def gerar_numero():
    return random.randint(1, 10)

def verificar_palpite(palpite, secreto):
    if palpite < secreto:
        return "maior"
    elif palpite > secreto:
        return "menor"
    else:
        return "acertou"

secreto = gerar_numero()
print("Número secreto gerado.")
print(verificar_palpite(5, secreto))
````

Separar em funções ajuda muito na leitura e também na correção de erros.

## 27. Comparando estilos de solução

Às vezes duas funções resolvem o mesmo problema, mas uma fica mais organizada.

### Versão menos organizada

````python
def resultado_aluno(n1, n2, n3):
    soma = n1 + n2 + n3
    media = soma / 3
    if media >= 7:
        return "Aprovado"
    else:
        return "Reprovado"

print(resultado_aluno(7, 8, 9))
````

### Versão com uma função auxiliar

````python
def calcular_media(n1, n2, n3):
    return (n1 + n2 + n3) / 3

def resultado_aluno(n1, n2, n3):
    media = calcular_media(n1, n2, n3)
    if media >= 7:
        return "Aprovado"
    else:
        return "Reprovado"

print(resultado_aluno(7, 8, 9))
````

A segunda abordagem costuma escalar melhor quando o programa cresce.

## 28. Quando criar uma função?

Uma boa regra prática é:

crie uma função quando:

- uma lógica vai se repetir;
- uma tarefa pode ser isolada;
- um bloco está ficando grande demais;
- você quer tornar o código mais legível.

Nem tudo precisa virar função imediatamente, mas muitas partes se beneficiam disso.
