# Aula: `while`, `break` e `continue`

## Objetivos
- Entender o laço `while`
- Usar `break` para parar um laço
- Usar `continue` para pular para a próxima iteração


## 1. Exemplo simples de while

```python
contador = 1
while contador <= 5:
    print("Contador:", contador)
    contador += 1
print("Fim do laço")
```

## 2. While com condição mudando

```python
senha_correta = "python123"
tentativas = ["aaa", "123", "python123"]
i = 0

while i < len(tentativas):
    tentativa = tentativas[i]
    print("Tentando:", tentativa)

    if tentativa == senha_correta:
        print("Acertou a senha!")
        break

    i += 1
```

## 3. Usando break

```python
numeros = [5, 8, 12, 20, 30]
alvo = 20
i = 0

while i < len(numeros):
    n = numeros[i]
    print("Analisando:", n)

    if n == alvo:
        print("Achei o número:", alvo)
        break

    i += 1
```

## 4. Usando continue

```python
n = 0
while n < 10:
    n += 1
    if n % 2 == 0:
        continue
    print("Ímpar:", n)
```

## 5. Exemplo final combinando tudo

```python
nomes = ["Ana", "", "Bruno", "FIM", "Carla"]
i = 0

while i < len(nomes):
    nome = nomes[i]
    i += 1

    if nome == "":
        continue

    if nome == "FIM":
        print("Encontrado 'FIM', encerrando...")
        break

    print("Nome:", nome)
```
