## 1. Por que usar laços?

Muitas vezes queremos repetir uma ação várias vezes.

Por exemplo, imagine mostrar os números de 1 até 5.

Sem laço, teríamos algo como:

```python
print(1)
print(2)
print(3)
print(4)
print(5)
```

Isso funciona, mas fica ruim quando a repetição é maior ou depende de dados.

Para isso, usamos **laços de repetição**.

## 2. O laço `for`

O `for` é usado quando queremos repetir algo percorrendo uma sequência.

Forma geral:

```python
for variavel in sequencia:
    # bloco de código
```

A cada repetição, a variável recebe um valor da sequência.

### Exemplo básico

````python
for i in [1, 2, 3, 4, 5]:
    print(i)
````

## 3. Usando `range()`

A função `range()` é muito usada com `for`.

Ela gera uma sequência de números inteiros.

### Forma 1: `range(fim)`

Vai de `0` até `fim - 1`.

````python
for i in range(5):
    print(i)
````

No exemplo acima, o resultado é:

```python
0
1
2
3
4
```

### Forma 2: `range(inicio, fim)`

Vai de `inicio` até `fim - 1`.

````python
for i in range(2, 7):
    print(i)
````

### Forma 3: `range(inicio, fim, passo)`

Permite controlar de quanto em quanto a sequência avança.

````python
for i in range(0, 11, 2):
    print(i)
````

Também podemos usar passo negativo.

````python
for i in range(10, 0, -1):
    print(i)
````

## 4. `for i in range(5)`

Esse é um dos formatos mais comuns.

Exemplo: repetir uma mensagem 5 vezes.

````python
for i in range(5):
    print('Olá!')
````

Também podemos usar `i` dentro do código.

````python
for i in range(5):
    print('Repetição número', i)
````

## 5. `for elemento in lista`

Quando queremos percorrer diretamente os elementos de uma lista, podemos fazer:

```python
for nome in nomes:
    print(nome)
```

````python
nomes = ['Ana', 'Bruno', 'Carlos']

for nome in nomes:
    print(nome)
````

Esse formato é muito bom quando queremos trabalhar diretamente com cada elemento.

## 6. `for i in range(len(lista))`

Às vezes não queremos apenas o elemento, mas também sua **posição**.

Nesse caso, usamos `range(len(lista))`.

````python
nomes = ['Ana', 'Bruno', 'Carlos']

for i in range(len(nomes)):
    print(i, nomes[i])
````

Esse padrão é útil quando precisamos acessar o índice da lista.

## 7. Comparando dois estilos de `for`

### Percorrendo elementos diretamente

````python
nums = [10, 20, 30]

for n in nums:
    print(n)
````

### Percorrendo por índice

````python
nums = [10, 20, 30]

for i in range(len(nums)):
    print(i, nums[i])
````

## 8. O laço `while`

O `while` repete um bloco **enquanto uma condição for verdadeira**.

Forma geral:

```python
while condicao:
    # bloco de código
```

### Exemplo básico

````python
i = 0

while i < 5:
    print(i)
    i += 1
````

É muito importante que a condição do `while` eventualmente deixe de ser verdadeira. Senão, teremos um **loop infinito**.

## 9. Exemplo de `while` com contador

Uma estrutura muito comum é usar uma variável contadora.

````python
contador = 1

while contador <= 5:
    print('Valor:', contador)
    contador += 1
````

## 10. Diferença entre `for` e `while`

### Use `for` quando:
- você souber quantas repetições quer fazer
- estiver percorrendo uma lista
- estiver usando `range()`

### Use `while` quando:
- a repetição depender de uma condição
- você não souber exatamente quantas vezes vai repetir
- quiser parar apenas quando algo acontecer

## 11. Exemplos com strings, listas e condicionais

Agora vamos ver aplicações mais próximas do que normalmente aparece em exercícios.

### Exemplo 1 — Mostrar os caracteres de uma string com `for`

````python
palavra = 'python'

for letra in palavra:
    print(letra)
````

### Exemplo 2 — Somar valores de uma lista

````python
nums = [10, 20, 30, 40]
soma = 0

for n in nums:
    soma += n

print('Soma =', soma)
````

### Exemplo 3 — Contar números pares em uma lista

````python
nums = [4, 7, 10, 13, 16, 21]
qtd_pares = 0

for n in nums:
    if n % 2 == 0:
        qtd_pares += 1

print('Quantidade de pares =', qtd_pares)
````

### Exemplo 4 — Percorrendo uma lista com índice

````python
nomes = ['Ana', 'Bruno', 'Carlos']

for i in range(len(nomes)):
    print('Posição', i, ':', nomes[i])
````

### Exemplo 5 — Mostrar números até o usuário acertar (`while`)

````python
numero = 3
palpite = 0

while palpite != numero:
    palpite = int(input('Tente adivinhar o número: '))

print('Acertou!')
````

### Exemplo 6 — Criar lista de dobrados com `for`

````python
nums = [1, 2, 3, 4]
dobrados = []

for n in nums:
    dobrados.append(n * 2)

print(dobrados)
````

### Exemplo 7 — Copiar lista com `while`

````python
nomes = ['Ana', 'Bruno', 'Carlos']
copia = []
i = 0

while i < len(nomes):
    copia.append(nomes[i])
    i += 1

print(copia)
````

## 12. Cuidados comuns

- lembrar que `range(5)` vai até 4
- não esquecer de atualizar a variável no `while`
- não confundir percorrer elementos com percorrer índices
- cuidado com `len(lista)` e posições válidas
- testar com exemplos pequenos

## 13. Boas práticas iniciais

- use nomes claros como `contador`, `soma`, `nome`, `nums`
- quando usar `while`, verifique como o laço vai terminar
- quando usar `for`, pense se você precisa do elemento ou do índice
- teste o código com poucos valores antes

## Exercício 14 — Filtrar números positivos

Dada a lista:

```python
nums = [5, -3, 8, -1, 7, -4]
```

Crie uma nova lista contendo apenas os números positivos.

````python
# Gabarito Exercício 14
nums = [5, -3, 8, -1, 7, -4]
positivos = []

for n in nums:
    if n > 0:
        positivos.append(n)

print(positivos)
````

## Exercício 15 — Intercalar listas

Dadas as listas:

```python

b = [10, 20, 30]
```

Crie uma nova lista intercalando os elementos.

````python
# Gabarito Exercício 15
a = [1, 2, 3]
b = [10, 20, 30]
c = []

for i in range(len(a)):
    c.append(a[i])
    c.append(b[i])

print(c)
````

## Exercício 16 — Inverter lista manualmente

Dada a lista:

```python
nums = [1, 2, 3, 4, 5]
```

Crie uma nova lista com os elementos em ordem inversa, sem usar `reverse()`.

````python
# Gabarito Exercício 16
nums = [1, 2, 3, 4, 5]
invertida = []

for i in range(len(nums) - 1, -1, -1):
    invertida.append(nums[i])

print(invertida)
````

## Exercício 19 — Número mais frequente sem dicionário

Dada a lista:

```python
nums = [1, 2, 2, 3, 3, 3, 4]
```

Retorne o número que aparece mais vezes, sem usar dicionário.

````python
# Gabarito Exercício 19
nums = [1, 2, 2, 3, 3, 3, 4]

nums.sort()

mais_frequente = nums[0]
maior_contagem = 1

contagem_atual = 1
anterior = nums[0]

for valor in nums[1:]:
    if valor == anterior:
        contagem_atual += 1
    else:
        if contagem_atual > maior_contagem:
            maior_contagem = contagem_atual
            mais_frequente = anterior
        anterior = valor
        contagem_atual = 1

if contagem_atual > maior_contagem:
    mais_frequente = anterior

print(mais_frequente)
````

## Exercício 20 — Dividir pares e ímpares

Dada a lista:

```python
nums = [3, 8, 11, 4, 6, 9]
```

Crie duas listas: uma com os pares e outra com os ímpares.

````python
# Gabarito Exercício 20
nums = [3, 8, 11, 4, 6, 9]

pares = []
impares = []

for n in nums:
    if n % 2 == 0:
        pares.append(n)
    else:
        impares.append(n)

print('Pares:', pares)
print('Ímpares:', impares)
````
