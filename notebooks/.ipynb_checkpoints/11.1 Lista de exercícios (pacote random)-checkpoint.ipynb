## Exercício 1 — Adivinhação com dicas

O computador escolhe um número aleatório entre **1 e 50**.

O jogador tenta adivinhar.

A cada tentativa, o programa informa se o palpite foi:

- `Muito alto`
- `Muito baixo`

O jogo termina quando o jogador acerta.

No final, mostre também **quantas tentativas foram necessárias**.

````python
# Gabarito Exercício 1
import random

numero = random.randint(1, 50)
tentativas = 0
palpite = 0

while palpite != numero:
    palpite = int(input('Digite seu palpite: '))
    tentativas += 1

    if palpite > numero:
        print('Muito alto')
    elif palpite < numero:
        print('Muito baixo')

print('Acertou!')
print('Tentativas:', tentativas)
````

## Exercício 2 — Pedra, papel ou tesoura

O usuário escolhe:

- `pedra`
- `papel`
- `tesoura`

O computador escolhe aleatoriamente.

O programa deve mostrar a jogada do usuário, a jogada do computador e o resultado final:

- vitória
- derrota
- empate

````python
# Gabarito Exercício 2
import random

opcoes = ['pedra', 'papel', 'tesoura']

usuario = input('Escolha pedra, papel ou tesoura: ').lower()
computador = random.choice(opcoes)

print('Você:', usuario)
print('Computador:', computador)

if usuario == computador:
    print('Empate!')
elif (usuario == 'pedra' and computador == 'tesoura') or \
     (usuario == 'papel' and computador == 'pedra') or \
     (usuario == 'tesoura' and computador == 'papel'):
    print('Você venceu!')
else:
    print('Você perdeu!')
````

## Exercício 3 — Gerador de times

Dada uma lista de jogadores, embaralhe a lista e divida os jogadores em **dois times com a mesma quantidade de pessoas**.

Exemplo de lista:

```python
jogadores = [
    'Ana', 'Bruno', 'Carlos', 'Daniel',
    'Eduardo', 'Fernanda', 'Gustavo', 'Helena'
]
```

````python
# Gabarito Exercício 3
import random

jogadores = [
    'Ana', 'Bruno', 'Carlos', 'Daniel',
    'Eduardo', 'Fernanda', 'Gustavo', 'Helena'
]

random.shuffle(jogadores)

metade = len(jogadores) // 2
time1 = jogadores[:metade]
time2 = jogadores[metade:]

print('Time 1:', time1)
print('Time 2:', time2)
````

## Exercício 4 — Palavra embaralhada

O computador escolhe uma palavra aleatória de uma lista.
Depois, embaralha as letras dessa palavra e mostra o resultado.

O jogador deve tentar adivinhar qual era a palavra original.

Exemplo de lista:

```python
palavras = ['python', 'computador', 'programacao', 'algoritmo']
```

````python
# Gabarito Exercício 4
import random

palavras = ['python', 'computador', 'programacao', 'algoritmo']

palavra = random.choice(palavras)
letras = list(palavra)
random.shuffle(letras)
embaralhada = ''.join(letras)

print('Palavra embaralhada:', embaralhada)

palpite = input('Qual é a palavra original? ').lower()

if palpite == palavra:
    print('Acertou!')
else:
    print('Errou! A palavra era', palavra)
````

## Exercício 5 — Loteria simplificada

O programa gera **5 números aleatórios diferentes entre 1 e 20**.

O usuário digita **5 números**.

No final, o programa mostra:

- os números sorteados
- quantos números o usuário acertou

````python
# Gabarito Exercício 5
import random

sorteados = random.sample(range(1, 21), 5)

usuario = []
for i in range(5):
    n = int(input(f'Digite o {i+1}º número: '))
    usuario.append(n)

acertos = 0
for n in usuario:
    if n in sorteados:
        acertos += 1

print('Números sorteados:', sorteados)
print('Seus números:', usuario)
print('Quantidade de acertos:', acertos)
````

## Exercício 6 — Gerador de senha

Crie um gerador de senha aleatória com:

- letras minúsculas
- letras maiúsculas
- números
- alguns símbolos

O usuário deve escolher o **tamanho da senha**.

````python
# Gabarito Exercício 6
import random

chars = 'abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789!@#$%&*'
tamanho = int(input('Digite o tamanho da senha: '))

senha = ''
for i in range(tamanho):
    senha += random.choice(chars)

print('Senha gerada:', senha)
````

## Exercício 7 — Batalha de dados

Simule um jogo em que:

- o jogador lança **2 dados**
- o computador lança **2 dados**

Quem tiver a maior soma vence.

O programa deve mostrar os valores obtidos e o vencedor.

````python
# Gabarito Exercício 7
import random

j1 = random.randint(1, 6)
j2 = random.randint(1, 6)
c1 = random.randint(1, 6)
c2 = random.randint(1, 6)

soma_jogador = j1 + j2
soma_computador = c1 + c2

print('Você:', j1, '+', j2, '=', soma_jogador)
print('Computador:', c1, '+', c2, '=', soma_computador)

if soma_jogador > soma_computador:
    print('Você venceu!')
elif soma_computador > soma_jogador:
    print('Computador venceu!')
else:
    print('Empate!')
````

## Exercício 8 — Caça ao tesouro

O programa escolhe um número aleatório entre **1 e 100**.

O jogador tem **7 tentativas** para acertar.

Após cada tentativa, o programa deve dizer:

- `Mais alto`
- `Mais baixo`

Se acertar, o jogo termina com sucesso.
Se acabar as tentativas, o programa mostra `Game over`.

````python
# Gabarito Exercício 8
import random

tesouro = random.randint(1, 100)
tentativa = 1
acertou = False

while tentativa <= 7 and acertou == False:
    palpite = int(input(f'Tentativa {tentativa}: '))

    if palpite == tesouro:
        acertou = True
        print('Você encontrou o tesouro!')
    elif palpite < tesouro:
        print('Mais alto')
    else:
        print('Mais baixo')

    tentativa += 1

if acertou == False:
    print('Game over')
    print('O número era', tesouro)
````

## Exercício 9 — Forca simplificada com `random`

O computador escolhe uma palavra aleatória.

O jogador tenta adivinhar **letra por letra**.

O programa deve mostrar o progresso da palavra, por exemplo:

```python
_ y _ _ o _
```

O jogo termina quando:

- a palavra é descoberta
- ou o jogador erra 6 vezes

````python
# Gabarito Exercício 9
import random

palavras = ['python', 'banana', 'computador', 'janela', 'algoritmo']
palavra = random.choice(palavras)

descoberta = ['_'] * len(palavra)
erros = 0

while '_' in descoberta and erros < 6:
    print('Palavra:', ' '.join(descoberta))
    letra = input('Digite uma letra: ').lower()

    if letra in palavra:
        for i in range(len(palavra)):
            if palavra[i] == letra:
                descoberta[i] = letra
    else:
        erros += 1
        print('Errou! Total de erros:', erros)

if '_' not in descoberta:
    print('Parabéns! Você descobriu a palavra:', palavra)
else:
    print('Você perdeu! A palavra era:', palavra)
````

## Exercício 10 — Roleta de números

O programa gera números aleatórios entre **1 e 10** até sair o número **7**.

Mostre todos os números gerados e, no final, quantos sorteios foram necessários.

````python
# Gabarito Exercício 10
import random

contador = 0
numero = 0

while numero != 7:
    numero = random.randint(1, 10)
    print(numero)
    contador += 1

print('Foram necessários', contador, 'sorteios')
````
