## Exercício 1 — Contar vogais

```
frase = input("Digite uma frase: ")

qtd_a = frase.count("a")
qtd_e = frase.count("e")
qtd_i = frase.count("i")
qtd_o = frase.count("o")
qtd_u = frase.count("u")

total_vogais = qtd_a + qtd_e + qtd_i + qtd_o + qtd_u
print(total_vogais)
```

## Exercício 2 — Contar palavra

```
frase = input("Frase: ")
palavra = input("Palavra: ")
print(frase.count(palavra))
```

## Exercício 3 — Palavra está na frase

```
frase = input("Frase: ")
palavra = input("Palavra: ")
print(palavra in frase)
```

## Exercício 4 — Substituição

```
frase = input("Frase: ")
nova = frase.replace("bolo","torta")
print(nova)
```

## Exercício 8 — Caracteres sem espaço

```
frase = input("Frase: ")
print(len(frase) - frase.count(" "))
```

## Exercício 9 — Letras 'a' e proporção

```
texto = input("Texto: ")
qtd = texto.count("a")
prop = qtd / len(texto)
print(qtd, prop)
```

## Exercício 10 — Telefone formatado

```
t = input("Telefone 11 dígitos: ")
print("(" + t[0:2] + ") " + t[2:7] + "-" + t[7:11])
```

## Exercício 11 — Data formatada

```
data = input("DDMMAAAA: ")
print(data[0:2] + "/" + data[2:4] + "/" + data[4:8])
```

## Exercício 12 — Usuário e domínio

```
email = input("E-mail: ")
pos = email.find("@")
print("Usuário:", email[:pos])
print("Domínio:", email[pos+1:])
```

## Exercício 13 — Remover vogais

```
txt = input("Texto: ")
out = txt.replace("a","").replace("e","").replace("i","").replace("o","").replace("u","")
print(out)
```

## Exercício 14 — Bordão

```
p = input("Palavra: ")
print(p[0] + "*"*(len(p)-2) + p[-1])
```

## Exercício 15 — Frase abreviada

```
fr = input("Frase 3 palavras: ")
esp1 = fr.find(" ")
resto = fr[esp1+1:]
esp2 = resto.find(" ")
p1 = fr[:esp1]
meio = resto[:esp2]
p3 = resto[esp2+1:]
print(p1 + " (" + meio[0] + ".) " + p3)
```
