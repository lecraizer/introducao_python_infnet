Resoluções completas para os 10 exercícios.

## Exercício 1 — Contar vogais até encontrar `"x"`
**Exemplo:** entrada: `programaxao` → saída: `4`

```
s = "programaxao"
vogais = "aeiouAEIOU"
cont = 0

for letra in s:
    if letra == "x":
        break
    if letra in vogais:
        cont += 1

cont
```

## Exercício 2 — Somar apenas números positivos até o primeiro negativo
**Exemplo:** `[3, 7, 0, 12, -5, 9]` → `22`

```
nums = [3, 7, 0, 12, -5, 9]
soma = 0

for n in nums:
    if n < 0:
        break
    soma += n

soma
```

## Exercício 3 — Lista com palavras de tamanho par
**Exemplo:** `python eh muito massa` → `['python','massa']`

```
frase = "python eh muito massa"
resultado = []
palavras = frase.split()

for p in palavras:
    if len(p) % 2 == 0:
        resultado.append(p)

resultado
```

## Exercício 4 — Verificar se a lista está em ordem estritamente crescente

```
nums = [1, 2, 3, 5, 8]
crescente = True

for i in range(len(nums)-1):
    if nums[i] >= nums[i+1]:
        crescente = False
        break

crescente
```

## Exercício 5 — Procurar um número usando `while` e `break`
Exemplos incluídos no código.

```
lista = [10, 5, 8, 20, 3]
alvo = 20
i = 0
encontrou = False

while i < len(lista):
    if lista[i] == alvo:
        print(f"Número {alvo} encontrado na posição {i}")
        encontrou = True
        break
    i += 1

if not encontrou:
    print(f"Número {alvo} não encontrado")
```

## Exercício 6 — Contar palavras que começam com a mesma letra da primeira

```
palavras = ["python", "pizza", "Data", "piano", "Java"]
base = palavras[0][0].lower()
cont = 0

for p in palavras:
    if p.lower().startswith(base):
        cont += 1

cont
```

## Exercício 7 — Alternar maiúsculas e minúsculas

```
s = "programar"
nova = ""

for i, letra in enumerate(s):
    if i % 2 == 0:
        nova += letra.upper()
    else:
        nova += letra.lower()

nova
```

## Exercício 8 — Remover palavras com menos de 4 letras usando `while`

```
palavras = ["eu", "gosto", "de", "python", "e", "css"]
i = 0

while i < len(palavras):
    if len(palavras[i]) < 4:
        palavras.pop(i)
    else:
        i += 1

palavras
```

## Exercício 9 — Encontrar o número mais frequente

```
nums = [1,2,2,3,3,3,4]
freq = {}

for n in nums:
    freq[n] = freq.get(n, 0) + 1

mais_freq = None
maior = -1

for k,v in freq.items():
    if v > maior:
        maior = v
        mais_freq = k

mais_freq
```

## Exercício 10 — Verificar anagramas

```
a = "roma"
b = "amor"

def eh_anagrama(x, y):
    return sorted(x) == sorted(y)

eh_anagrama(a, b)
```
