# Lista de Exercícios – Strings e `input()`

Neste notebook você encontrará **10 exercícios** sobre o uso de **strings** e da função `input()` em Python. Os exercícios estão organizados em três níveis de dificuldade:

- 🟢 3 fáceis
- 🟡 4 médios
- 🔴 3 difíceis

Cada exercício vem seguido de um possível **gabarito** em uma célula de código.

## Exercício 1 (fácil) – Segundos em minutos e segundos
Peça ao usuário para digitar uma quantidade **inteira de segundos**. Depois, calcule quantos **minutos** inteiros cabem nessa quantidade e quantos **segundos** sobram. Mostre uma mensagem no formato, por exemplo:

`130 segundos são 2 minutos e 10 segundos.`

```
# Gabarito Exercício 1
segundos_texto = input('Digite uma quantidade de segundos: ')
segundos = int(segundos_texto)

minutos = segundos // 60
resto_segundos = segundos % 60

mensagem = (
    str(segundos) + ' segundos são ' +
    str(minutos) + ' minutos e ' +
    str(resto_segundos) + ' segundos.'
)

print(mensagem)
```

## Exercício 2 (fácil) – Apresentação personalizada
Peça ao usuário para digitar seu **nome** e a **cidade** onde mora. Depois, mostre uma frase no formato:

`Olá, NOME! Você mora em CIDADE.`

```
# Gabarito Exercício 2
nome = input('Digite seu nome: ')
cidade = input('Digite a cidade onde você mora: ')

frase = 'Olá, ' + nome + '! Você mora em ' + cidade + '.'
print(frase)
```

## Exercício 3 (fácil) – Contando letras e caracteres
Peça ao usuário para digitar uma **frase** qualquer. Depois mostre:

- A quantidade total de caracteres na frase;
- Quantas vezes a letra `'a'` aparece na frase (maiúscula ou minúscula).


```
# Gabarito Exercício 3
frase = input('Digite uma frase: ')

qtd_caracteres = len(frase)
qtd_a = frase.lower().count('a')

print('Total de caracteres:', qtd_caracteres)
print("Quantidade de 'a' (maiúscula ou minúscula):", qtd_a)
```

## Exercício 4 (médio) – Nome completo formatado
Peça ao usuário para digitar seu **primeiro nome** e depois seu **sobrenome**. Monte uma string no formato `'SOBRENOME, Nome'`, em que o sobrenome aparece todo em letras maiúsculas e o primeiro nome começa com letra maiúscula e o restante minúsculas. Mostre o resultado na tela.

```
# Gabarito Exercício 4
primeiro_nome = input('Digite seu primeiro nome: ')
sobrenome = input('Digite seu sobrenome: ')

primeiro_formatado = primeiro_nome.capitalize()
sobrenome_formatado = sobrenome.upper()

resultado = sobrenome_formatado + ', ' + primeiro_formatado
print(resultado)
```

## Exercício 5 (médio) – Procurando uma palavra na frase
Peça ao usuário para digitar uma **frase** e depois uma **palavra**. Verifique se a palavra aparece na frase. Mostre:

- `True` ou `False` indicando se a palavra aparece;
- Quantas vezes a palavra aparece dentro da frase.


```
# Gabarito Exercício 5
frase = input('Digite uma frase: ')
palavra = input('Digite uma palavra para buscar: ')

# para uma busca mais simples, vamos comparar tudo em minúsculas
frase_lower = frase.lower()
palavra_lower = palavra.lower()

aparece = palavra_lower in frase_lower
qtd = frase_lower.count(palavra_lower)

print('A palavra aparece?', aparece)
print('Quantidade de ocorrências:', qtd)
```

## Exercício 6 (médio) – Transformações de texto
Peça ao usuário para digitar uma **frase**. Mostre na tela:

- A frase original;
- A frase toda em letras maiúsculas;
- A frase toda em letras minúsculas;
- A frase com a primeira letra de cada palavra em maiúscula.


```
# Gabarito Exercício 6
frase = input('Digite uma frase: ')

print('Original:         ', frase)
print('Maiúsculas:       ', frase.upper())
print('Minúsculas:       ', frase.lower())
print('Título (Title):   ', frase.title())
```

## Exercício 7 (médio) – Cálculo simples com conversão
Peça ao usuário para digitar seu **peso em kg** e sua **altura em metros**. Leia tudo como texto, converta para `float` e calcule o IMC (índice de massa corporal) usando a fórmula:

\[ IMC = \frac{peso}{altura^2} \]

Mostre uma mensagem no formato:

`Seu IMC é X.XX`

(com o valor arredondado para duas casas decimais).

```
# Gabarito Exercício 7
peso_texto = input('Digite seu peso em kg: ')
altura_texto = input('Digite sua altura em metros (ex: 1.75): ')

peso = float(peso_texto)
altura = float(altura_texto)

imc = peso / (altura ** 2)

mensagem = 'Seu IMC é ' + f'{imc:.2f}'
print(mensagem)
```

## Exercício 8 (difícil) – Analisando um e-mail
Peça ao usuário para digitar um **endereço de e-mail**. Faça uma verificação simples:

- O texto deve conter o caractere `'@'`;
- Após o `'@'`, deve existir pelo menos um ponto `'.'`.

Se essas condições forem satisfeitas, mostre:

- A parte **antes** do `'@'` (nome do usuário);
- A parte **depois** do `'@'` (domínio completo).

Se não forem satisfeitas, mostre uma mensagem dizendo que o e-mail é inválido.

```
# Gabarito Exercício 8
email = input('Digite um endereço de e-mail: ')

if '@' in email:
    pos_arroba = email.index('@')
    usuario = email[:pos_arroba]
    dominio = email[pos_arroba + 1:]

    if '.' in dominio and len(usuario) > 0 and len(dominio) > 0:
        print('Usuário:', usuario)
        print('Domínio:', dominio)
    else:
        print('E-mail inválido (problema no domínio).')
else:
    print('E-mail inválido (não contém @).')
```

## Exercício 9 (difícil) – Estatísticas de uma frase
Peça ao usuário para digitar uma **frase**. Em seguida, calcule e mostre:

- O número total de caracteres;
- O número de palavras (considere que as palavras estão separadas por espaços);
- A **primeira palavra** e a **última palavra**;
- A média aproximada de caracteres por palavra (use o total de caracteres sem contar os espaços, dividido pelo número de palavras).


```
# Gabarito Exercício 9
frase = input('Digite uma frase: ')

qtd_caracteres = len(frase)

# separando por espaços
palavras = frase.split()
qtd_palavras = len(palavras)

if qtd_palavras > 0:
    primeira = palavras[0]
    ultima = palavras[-1]

    # contando caracteres sem espaços
    frase_sem_espacos = frase.replace(' ', '')
    qtd_caracteres_sem_espacos = len(frase_sem_espacos)

    media = qtd_caracteres_sem_espacos / qtd_palavras

    print('Total de caracteres (com espaços):', qtd_caracteres)
    print('Número de palavras:', qtd_palavras)
    print('Primeira palavra:', primeira)
    print('Última palavra:', ultima)
    print('Média de caracteres por palavra:', f'{media:.2f}')
else:
    print('Nenhuma palavra foi digitada.')
```

## Exercício 10 (difícil) – Substituindo palavras em uma frase
Peça ao usuário para digitar uma **frase**. Em seguida, peça duas palavras:

- Uma palavra que será **procurada** na frase;
- Uma palavra que será usada como **substituta**.

Mostre na tela:

- A frase original;
- A frase depois da substituição de todas as ocorrências;
- Quantas substituições foram feitas.


```
# Gabarito Exercício 10
frase = input('Digite uma frase: ')
palavra_antiga = input('Digite a palavra que será substituída: ')
palavra_nova = input('Digite a nova palavra: ')

# vamos contar quantas vezes a palavra antiga aparece (busca exata, sensível a maiúsculas/minúsculas)
qtd_substituicoes = frase.count(palavra_antiga)

frase_substituida = frase.replace(palavra_antiga, palavra_nova)

print('Frase original:    ', frase)
print('Frase modificada:  ', frase_substituida)
print('Substituições feitas:', qtd_substituicoes)
```
