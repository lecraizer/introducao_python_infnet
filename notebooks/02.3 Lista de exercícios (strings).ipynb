## Exercício 1 - Contar vogais

```python
frase = "python é muito legal"

qtd_a = frase.count("a")
qtd_e = frase.count("e")
qtd_i = frase.count("i")
qtd_o = frase.count("o")
qtd_u = frase.count("u")

total_vogais = qtd_a + qtd_e + qtd_i + qtd_o + qtd_u
print(total_vogais)
```

## Exercício 9 - Letras 'a' e proporção

```python
texto = "a casa amarela na praia"
qtd_a = texto.count("a")
tamanho = len(texto)
proporcao = qtd_a / tamanho
print("Quantidade de 'a':", qtd_a)
print("Proporção:", proporcao)
```

## Exercício 10 - Telefone formatado

```python
telefone = "21987654321"

ddd = telefone[0:2]
parte1 = telefone[2:7]
parte2 = telefone[7:11]

formatado = "(" + ddd + ") " + parte1 + "-" + parte2
print(formatado)
```

## Exercício 11 - Data formatada

```python
data = "25032024"

dia = data[0:2]
mes = data[2:4]
ano = data[4:8]

print(dia + "/" + mes + "/" + ano)
```

## Exercício 12 - Usuário e domínio

```python
email = "aluno123@gmail.com"

pos = email.find("@")
usuario = email[:pos]
dominio = email[pos+1:]

print("Usuário:", usuario)
print("Domínio:", dominio)
```

## Exercício 14 - Bordão

```python
palavra = input('Digite uma palavra qualquer: ')

primeira = palavra[0]
ultima = palavra[-1]
meio = "*" * (len(palavra) - 2)

resultado = primeira + meio + ultima
print(resultado)
```

## Exercício 15 - Frase abreviada

```python
frase = input('Digite uma frase qualquer: ')

esp1 = frase.find(" ")
resto = frase[esp1 + 1:]
esp2 = resto.find(" ")

p1 = frase[:esp1]
meio = resto[:esp2]
p3 = resto[esp2 + 1:]

abreviado = p1 + " (" + meio[0] + ".) " + p3
print(abreviado)
```
