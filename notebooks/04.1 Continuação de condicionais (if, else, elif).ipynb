## 1. O que são condicionais?

Em programação, uma **condicional** permite que o programa tome decisões.

Ou seja: dependendo de uma condição ser **verdadeira** (`True`) ou **falsa** (`False`), um bloco de código será executado.

Exemplo do dia a dia:

- Se estiver chovendo, levo guarda-chuva.
- Senão, saio sem guarda-chuva.

Em Python, isso aparece com estruturas como:

```python
if condicao:
    # faz algo
else:
    # faz outra coisa
```

## 2. Fluxo de execução

O programa normalmente executa as linhas **de cima para baixo**.

As condicionais alteram esse caminho. Elas funcionam como um “desvio” no fluxo:

1. Python avalia a condição.
2. Se o resultado for `True`, entra no bloco do `if`.
3. Se for `False`, pula esse bloco.
4. Caso exista `else`, entra nele.

Veja a ideia:

```python
idade = 20

if idade >= 18:
    print("Maior de idade")
else:
    print("Menor de idade")

print("Fim do programa")
```

Se `idade >= 18` for verdadeira, o Python imprime `"Maior de idade"`.
Depois disso, ele continua o fluxo normal até `"Fim do programa"`.

## 3. Operadores relacionais

Esses operadores são usados para comparar valores.

| Operador | Significado | Exemplo |
|---|---|---|
| `==` | igual a | `x == 10` |
| `!=` | diferente de | `x != 10` |
| `>` | maior que | `x > 10` |
| `<` | menor que | `x < 10` |
| `>=` | maior ou igual a | `x >= 10` |
| `<=` | menor ou igual a | `x <= 10` |

Exemplo:
```python
nota = 7

if nota >= 6:
    print("Aprovado")
```

````python
nota = 7

if nota >= 6:
    print("Aprovado")
````

## 4. `if`

O `if` executa um bloco **somente se** a condição for verdadeira.

Sintaxe:

```python
if condicao:
    bloco
```

Exemplo:

````python
temperatura = 31

if temperatura > 30:
    print("Está quente")
````

Se a condição for falsa, o Python simplesmente **não entra** no bloco.

## 5. `else`

O `else` representa o “caso contrário”.

Sintaxe:

```python
if condicao:
    bloco_1
else:
    bloco_2
```

Exemplo:

````python
numero = 5

if numero % 2 == 0:
    print("Par")
else:
    print("Ímpar")
````

## 6. `elif`

O `elif` serve para testar **outras condições** quando o `if` inicial for falso.

Sintaxe:

```python
if condicao_1:
    bloco_1
elif condicao_2:
    bloco_2
elif condicao_3:
    bloco_3
else:
    bloco_final
```

Exemplo:

````python
nota = 8.5

if nota >= 9:
    print("Excelente")
elif nota >= 7:
    print("Bom")
elif nota >= 6:
    print("Aprovado")
else:
    print("Reprovado")
````

### Importante

O Python testa as condições **na ordem**.

Assim que encontra uma condição verdadeira, ele executa aquele bloco e **ignora os próximos**.

## 7. Indentação

Em Python, a indentação faz parte da sintaxe.

O bloco que pertence ao `if`, `elif` ou `else` deve ficar recuado, normalmente com 4 espaços.

Exemplo correto:

```python
idade = 18

if idade >= 18:
    print("Pode entrar")
else:
    print("Não pode entrar")
```

Se a indentação estiver errada, o código pode dar erro ou funcionar de forma inesperada.

## 8. Operadores lógicos

Às vezes uma única comparação não basta. Nesses casos usamos operadores lógicos.

### `and`
As duas condições precisam ser verdadeiras.

```python
idade = 20
tem_ingresso = True

if idade >= 18 and tem_ingresso:
    print("Entrada permitida")
```

### `or`
Basta uma das condições ser verdadeira.

```python
if dia == "sábado" or dia == "domingo":
    print("Fim de semana")
```

### `not`
Inverte o valor lógico.

```python
if not chovendo:
    print("Vou caminhar")
```

````python
idade = 20
tem_ingresso = True

if idade >= 18 and tem_ingresso:
    print("Entrada permitida")
else:
    print("Entrada negada")
````

````python
dia = "sábado"

if dia == "sábado" or dia == "domingo":
    print("Fim de semana")
else:
    print("Dia útil")
````

````python
chovendo = False

if not chovendo:
    print("Vou caminhar")
else:
    print("Melhor ficar em casa")
````

## 9. Condições compostas

Podemos combinar comparações para tomar decisões mais específicas.

Exemplo:

````python
idade = 25
renda = 3500

if idade >= 18 and renda >= 2000:
    print("Pode solicitar o cartão")
else:
    print("Não atende aos critérios")
````

## 10. Condicionais aninhadas

Também é possível colocar um `if` dentro de outro.

Exemplo:

````python
idade = 20
tem_documento = True

if idade >= 18:
    if tem_documento:
        print("Entrada liberada")
    else:
        print("Precisa apresentar documento")
else:
    print("Menor de idade")
````

Na prática, muitas vezes dá para reescrever isso com operadores lógicos, deixando o código mais enxuto.

## 11. Valores booleanos

O tipo `bool` em Python possui apenas dois valores:

- `True`
- `False`

Muitas expressões retornam booleanos:

````python
print(10 > 3)
print(5 == 5)
print(7 != 2)
print(4 <= 1)
````

## 12. Boas práticas

Algumas boas práticas ao escrever condicionais:

1. Prefira condições claras.
2. Evite comparações desnecessárias.
3. Teste os casos de borda.
4. Tome cuidado com a ordem dos `elif`.
5. Use nomes de variáveis significativos.

### Exemplo menos bom
```python
if aprovado == True:
    print("OK")
```

### Exemplo melhor
```python
if aprovado:
    print("OK")
```

## 13. Casos de borda

Casos de borda são valores que ficam “na fronteira” de uma regra.

Por exemplo, se a regra é `nota >= 6`, então a nota `6` é um caso importante para testar.

Outro exemplo:
- `idade >= 18`
- vale testar idade = 17, 18 e 19

Isso ajuda a evitar erros de lógica.

## 14. Resumo rápido

- `if`: executa algo se a condição for verdadeira
- `else`: executa no caso contrário
- `elif`: testa outras condições
- operadores relacionais comparam valores
- operadores lógicos combinam condições
- a ordem das condições importa
- a indentação é obrigatória em Python

## Exercício 1 — Verificar se a pessoa é maior de idade

Peça uma idade ao usuário e mostre:

- `Maior de idade`, se a idade for maior ou igual a 18
- `Menor de idade`, caso contrário

````python
idade = int(input("Digite sua idade: "))

if idade >= 18:
    print("Maior de idade")
else:
    print("Menor de idade")
````

## Exercício 2 — Verificar se um número é positivo

Peça um número ao usuário e mostre:

- `Positivo`, se ele for maior que zero
- `Zero ou negativo`, caso contrário

````python
numero = float(input("Digite um número: "))

if numero > 0:
    print("Positivo")
else:
    print("Zero ou negativo")
````

## Exercício 3 — Par ou ímpar

Peça um número inteiro e informe se ele é par ou ímpar.

````python
n = int(input("Digite um número inteiro: "))

if n % 2 == 0:
    print("Par")
else:
    print("Ímpar")
````

## Exercício 4 — Senha correta

Crie uma senha fixa, por exemplo `python123`.
Peça ao usuário uma senha e informe se ela está correta ou incorreta.

````python
senha_correta = "python123"
senha = input("Digite a senha: ")

if senha == senha_correta:
    print("Senha correta")
else:
    print("Senha incorreta")
````

## Exercício 5 — Classificação de nota

Peça uma nota ao usuário e mostre:

- `Excelente`, se a nota for maior ou igual a 9
- `Bom`, se a nota for maior ou igual a 7 e menor que 9
- `Aprovado`, se a nota for maior ou igual a 6 e menor que 7
- `Reprovado`, caso contrário

````python
nota = float(input("Digite a nota: "))

if nota >= 9:
    print("Excelente")
elif nota >= 7:
    print("Bom")
elif nota >= 6:
    print("Aprovado")
else:
    print("Reprovado")
````

## Exercício 6 — Pode votar?

Peça a idade da pessoa e informe:

- `Não vota`, se tiver menos de 16 anos
- `Voto facultativo`, se tiver 16 ou 17 anos, ou mais de 70
- `Voto obrigatório`, nos demais casos

````python
idade = int(input("Digite a idade: "))

if idade < 16:
    print("Não vota")
elif idade == 16 or idade == 17 or idade > 70:
    print("Voto facultativo")
else:
    print("Voto obrigatório")
````

## Exercício 7 — Acesso ao evento

Peça a idade e pergunte se a pessoa tem convite (`sim` ou `não`).
Mostre `Entrada permitida` apenas se ela for maior de idade **e** tiver convite.
Caso contrário, mostre `Entrada negada`.

````python
idade = int(input("Digite a idade: "))
convite = input("Tem convite? ").lower()

if idade >= 18 and convite == "sim":
    print("Entrada permitida")
else:
    print("Entrada negada")
````

## Exercício 8 — Menor entre dois números

Peça dois números e mostre qual deles é o menor.
Se forem iguais, informe isso.

````python
a = float(input("Digite o primeiro número: "))
b = float(input("Digite o segundo número: "))

if a < b:
    print("O menor número é", a)
elif b < a:
    print("O menor número é", b)
else:
    print("Os números são iguais")
````

## Exercício 9 — Calculadora simples

Peça dois números e uma operação (`+`, `-`, `*`, `/`).
Mostre o resultado da operação escolhida.
Se a operação for inválida, informe isso.
No caso da divisão, trate o caso em que o segundo número seja zero.

````python
a = float(input("Digite o primeiro número: "))
b = float(input("Digite o segundo número: "))
op = input("Digite a operação (+, -, *, /): ")

if op == "+":
    print("Resultado:", a + b)
elif op == "-":
    print("Resultado:", a - b)
elif op == "*":
    print("Resultado:", a * b)
elif op == "/":
    if b != 0:
        print("Resultado:", a / b)
    else:
        print("Não é possível dividir por zero")
else:
    print("Operação inválida")
````

## Exercício 10 — Ano bissexto

Peça um ano e informe se ele é bissexto.

Regra:
- é bissexto se for divisível por 4
- exceto os divisíveis por 100
- mas os divisíveis por 400 voltam a ser bissextos

````python
ano = int(input("Digite um ano: "))

if (ano % 4 == 0 and ano % 100 != 0) or (ano % 400 == 0):
    print("Ano bissexto")
else:
    print("Ano não bissexto")
````

## Exercício 11 — Triângulo válido

Peça três lados e verifique se eles podem formar um triângulo.

Dica matemática:
um triângulo só existe se cada lado for menor que a soma dos outros dois.

````python
a = float(input("Digite o lado 1: "))
b = float(input("Digite o lado 2: "))
c = float(input("Digite o lado 3: "))

if a < b + c and b < a + c and c < a + b:
    print("Forma um triângulo")
else:
    print("Não forma um triângulo")
````

## Exercício 12 — Classificação de triângulo

Depois de verificar se três lados formam um triângulo, classifique-o como:

- `Equilátero`: três lados iguais
- `Isósceles`: dois lados iguais
- `Escaleno`: três lados diferentes
- ou informe que não forma triângulo

````python
a = float(input("Digite o lado 1: "))
b = float(input("Digite o lado 2: "))
c = float(input("Digite o lado 3: "))

if a < b + c and b < a + c and c < a + b:
    if a == b and b == c:
        print("Triângulo equilátero")
    elif a == b or a == c or b == c:
        print("Triângulo isósceles")
    else:
        print("Triângulo escaleno")
else:
    print("Não forma triângulo")
````
