## Exercício 1 — Contar vogais até encontrar `"x"`

Dada uma string, percorra letra por letra e conte quantas vogais aparecem **antes** da letra `"x"`.  
Se aparecer `"x"`, pare imediatamente.

**Exemplo:**  
Entrada: `"programaxao"` → Saída esperada: `4`  
(vogais: o, a, a, a antes do x)


```python
s = "programaxao"
vogais = "aeiouAEIOU"
cont = 0

for letra in s:
    if letra == "x":
        break
    if letra in vogais:
        cont += 1

cont
```

## Exercício 2 — Somar números positivos até o primeiro negativo

Some todos os números **maiores que zero** até o primeiro número negativo.  
O negativo não entra na soma.

**Exemplo:**  
Entrada: `[3, 7, 0, 12, -5, 9]` → Saída esperada: `22`


```python
nums = [3, 7, 0, 12, -5, 9]
soma = 0

for n in nums:
    if n < 0:
        break
    soma += n

soma
```

## Exercício 3 — Lista com palavras de tamanho par

Crie uma lista contendo apenas as palavras cujo número de letras é **par**.

**Exemplo:**  
Entrada: `"python eh muito massa"` → Saída: `["python", "massa"]`


```python
frase = "python eh muito massa"
palavras = frase.split()
res = []

for p in palavras:
    if len(p) % 2 == 0:
        res.append(p)

res
```

## Exercício 4 — Verificar se lista é estritamente crescente

Retorne `True` se a lista for estritamente crescente; caso contrário, `False`.

**Exemplo 1:** `[1, 2, 3, 5, 8]` → `True`  
**Exemplo 2:** `[1, 2, 2, 3]` → `False`


```python
nums = [1, 2, 3, 5, 8]
cresc = True

for i in range(len(nums)-1):
    if nums[i] >= nums[i+1]:
        cresc = False
        break

cresc
```

## Exercício 5 — Procurar número usando `while` e `break`

Percorra uma lista até encontrar o número-alvo. Mostre a posição se encontrar; caso contrário, indique que não encontrou.

**Exemplo:**  
Lista `[10,5,8,20,3]`, alvo `20` → `"Número 20 encontrado na posição 3"`


```python
lista = [10, 5, 8, 20, 3]
alvo = 20
i = 0
found = False

while i < len(lista):
    if lista[i] == alvo:
        print(f"Número {alvo} encontrado na posição {i}")
        found = True
        break
    i += 1

if not found:
    print(f"Número {alvo} não encontrado")
```

## Exercício 6 — Contar palavras que começam com a mesma letra da primeira

Use a primeira palavra da lista como referência e conte quantas começam com a mesma letra.

**Exemplo:**  
Entrada: `["python", "pizza", "Data", "piano", "Java"]` → Saída esperada: `3`


```python
palavras = ["python", "pizza", "Data", "piano", "Java"]
base = palavras[0][0].lower()
cont = 0

for p in palavras:
    if p.lower().startswith(base):
        cont += 1

cont
```

## Exercício 7 — Alternar maiúsculas e minúsculas

Letras nas posições pares → maiúsculas  
Letras nas posições ímpares → minúsculas  

**Exemplo:** `"programar"` → `"PrOgRaMaR"`


```python
s = "programar"
nova = ""

for i, letra in enumerate(s):
    if i % 2 == 0:
        nova += letra.upper()
    else:
        nova += letra.lower()

nova
```

## Exercício 8 — Remover palavras com menos de 4 letras (while)

Use um `while` para ir removendo palavras menores que 4 letras até que só restem palavras ≥ 4.

**Exemplo:**  
Entrada: `["eu","gosto","de","python","e","css"]` → `["gosto","python"]`


```python
palavras = ["eu","gosto","de","python","e","css"]
i = 0

while i < len(palavras):
    if len(palavras[i]) < 4:
        palavras.pop(i)
    else:
        i += 1

palavras
```

## Exercício 9 — Encontrar o número mais frequente

Retorne o número que aparece mais vezes.

**Exemplo:** `[1,2,2,3,3,3,4]` → `3`


```python
nums = [1,2,2,3,3,3,4]

# ordena a lista para agrupar iguais
nums.sort()

mais_frequente = nums[0]
maior_contagem = 1

contagem_atual = 1
anterior = nums[0]

for valor in nums[1:]:
    if valor == anterior:
        contagem_atual += 1
    else:
        if contagem_atual > maior_contagem:
            maior_contagem = contagem_atual
            mais_frequente = anterior
        anterior = valor
        contagem_atual = 1

# checa o último grupo
if contagem_atual > maior_contagem:
    mais_frequente = anterior

mais_frequente
```
