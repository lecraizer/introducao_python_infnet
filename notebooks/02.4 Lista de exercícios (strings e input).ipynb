## Exercício 1 (fácil) – Segundos em minutos e segundos
Peça ao usuário para digitar uma quantidade inteira de segundos e converta para minutos e segundos.

```python
segundos = int(input('Digite os segundos: '))
minutos = segundos // 60
resto = segundos % 60
print(str(segundos) + ' segundos são ' + str(minutos) + ' minutos e ' + str(resto) + ' segundos.')
```

## Exercício 3 (fácil) – Contando letras e caracteres
Peça uma frase e conte:
- total de caracteres
- quantidade de 'a' (maiúsculo/minúsculo)

```python
frase = input('Digite uma frase: ')
print('Total de caracteres:', len(frase))
print("Qtd de 'a':", frase.lower().count('a'))
```

## Exercício 5 (médio) – Procurando palavra
Busque ocorrências de uma palavra dentro de uma frase usando `in` e `count()`.

```python
frase = input('Digite a frase: ')
pal = input('Digite a palavra: ')
f = frase.lower()
p = pal.lower()
print('Aparece?:', p in f)
print('Ocorrências:', f.count(p))
```

## Exercício 7 (médio) – Média de três notas
Peça três notas como strings, converta para float e calcule a média.

```python
disc = input('Disciplina: ')
n1 = float(input('Nota 1: '))
n2 = float(input('Nota 2: '))
n3 = float(input('Nota 3: '))
media = (n1 + n2 + n3) / 3
print('A média em ' + disc + ' é ' + f'{media:.2f}')
```

## Exercício 8 (difícil) – Separando partes do e-mail (assuma e-mail válido)
Peça um e-mail e separe a parte antes e depois do `'@'`, assumindo que o e-mail contém exatamente um `'@'`.

```python
email = input('Digite um e-mail: ')
pos = email.index('@')
print('Usuário:', email[:pos])
print('Domínio:', email[pos+1:])
```

## Exercício 9 (difícil) – Manipulação extensa de uma frase
Peça ao usuário uma frase. Depois mostre:

- A frase original;
- A frase sem espaços (use `replace`);
- Quantas vezes aparecem as vogais `'a'`, `'e'`, `'i'`, `'o'`, `'u'` (maiúsculas/minúsculas);
- A frase com todas as letras alternando caixa (use `swapcase()`).

```python
f = input('Digite uma frase: ')

sem_espacos = f.replace(' ', '')
fl = f.lower()

qta = fl.count('a')
qte = fl.count('e')
qti = fl.count('i')
qto = fl.count('o')
qtu = fl.count('u')

print('Frase original:', f)
print('Sem espaços:', sem_espacos)
print('Vogais: a=', qta, ', e=', qte, ', i=', qti, ', o=', qto, ', u=', qtu)
print('Swapcase:', f.swapcase())
```

## Exercício 10 (difícil) – Substituições
Peça uma frase e duas palavras (a antiga e a nova). Substitua todas as ocorrências e mostre quantas foram.

```python
fr = input('Digite uma frase: ')
old = input('Palavra antiga: ')
new = input('Palavra nova: ')
qtd = fr.count(old)
print('Original:', fr)
print('Modificada:', fr.replace(old, new))
print('Substituições:', qtd)
```
