# Exercícios de Python — Listas, `for` e `while`

Este notebook reúne exercícios focados em **listas**, **concatenação de listas**, **laços `for` e `while`**, além de construção de novas listas e análise de valores.

Cada exercício vem acompanhado de um possível **gabarito**.

## Exercício 1 — Soma dos elementos da lista
Dada a lista:

```python
nums = [3, 7, 2, 9, 4]
```

Calcule a **soma de todos os elementos** usando um `for`.

````python
# Gabarito Exercício 1
nums = [3, 7, 2, 9, 4]

soma = 0
for n in nums:
    soma += n

print(soma)
````

## Exercício 2 — Maior número da lista
Dada a lista:

```python
nums = [12, 5, 8, 21, 7]
```

Encontre o **maior valor da lista** sem usar `max()`.

````python
# Gabarito Exercício 2
nums = [12, 5, 8, 21, 7]

maior = nums[0]

for n in nums:
    if n > maior:
        maior = n

print(maior)
````

## Exercício 3 — Contar números pares
Dada a lista:

```python
nums = [4, 7, 10, 13, 16, 21]
```

Conte quantos números **pares** existem na lista.

````python
# Gabarito Exercício 3
nums = [4, 7, 10, 13, 16, 21]

qtd_pares = 0

for n in nums:
    if n % 2 == 0:
        qtd_pares += 1

print(qtd_pares)
````

## Exercício 9 — Intercalar listas


Crie uma nova lista **intercalando os elementos**.

Resultado esperado:


````python
# Gabarito Exercício 9
a = [1, 2, 3]
b = [10, 20, 30]

c = []

for i in range(len(a)):
    c.append(a[i])
    c.append(b[i])

print(c)
````

## Exercício 10 — Encontrar o segundo maior número
Dada a lista:

```python
nums = [10, 4, 7, 21, 15]
```

Encontre o **segundo maior valor** da lista **sem usar `sort()` ou `max()`**.

````python
# Gabarito Exercício 10
nums = [10, 4, 7, 21, 15]

maior = nums[0]
segundo_maior = nums[0]

for n in nums:
    if n > maior:
        segundo_maior = maior
        maior = n
    elif n > segundo_maior and n != maior:
        segundo_maior = n

print(segundo_maior)
````

## Exercício 11 — Inverter lista manualmente
Dada a lista:

```python
nums = [1, 2, 3, 4, 5]
```

Crie uma nova lista contendo os mesmos elementos **em ordem inversa**, usando `for`.

````python
# Gabarito Exercício 11
nums = [1, 2, 3, 4, 5]

invertida = []

for i in range(len(nums) - 1, -1, -1):
    invertida.append(nums[i])

print(invertida)
````

## Exercício 13 — Lista de números únicos
Dada a lista:

```python
nums = [1, 2, 2, 3, 3, 3, 4]
```

Crie uma nova lista contendo **apenas os números que aparecem pela primeira vez**, ignorando repetições posteriores.

Resultado esperado:

```python
[1, 2, 3, 4]
```

**Sem usar `set`.**

````python
# Gabarito Exercício 13
nums = [1, 2, 2, 3, 3, 3, 4]

unicos = []

for n in nums:
    if n not in unicos:
        unicos.append(n)

print(unicos)
````

## Exercício 14 — Dividir lista em pares e ímpares
Dada a lista:

```python
nums = [3, 8, 11, 4, 6, 9]
```

Crie duas listas: uma chamada `pares` e outra chamada `impares`.

````python
# Gabarito Exercício 14
nums = [3, 8, 11, 4, 6, 9]

pares = []
impares = []

for n in nums:
    if n % 2 == 0:
        pares.append(n)
    else:
        impares.append(n)

print('pares =', pares)
print('impares =', impares)
````

## Exercício 15 — Média da lista
Dada a lista:

```python
nums = [10, 20, 30, 40]
```

Calcule a **média dos valores** usando um loop.

````python
# Gabarito Exercício 15
nums = [10, 20, 30, 40]

soma = 0

for n in nums:
    soma += n

media = soma / len(nums)

print(media)
````

## Exercício 16 — Soma usando `while`
Dada a lista:

```python
nums = [2, 4, 6, 8]
```

Calcule a soma dos elementos usando `while` em vez de `for`.

````python
# Gabarito Exercício 16
nums = [2, 4, 6, 8]

i = 0
soma = 0

while i < len(nums):
    soma += nums[i]
    i += 1

print(soma)
````

## Exercício 17 — Contar quantos elementos são maiores que 10
Dada a lista:

```python
nums = [5, 12, 8, 20, 3, 17]
```

Conte quantos elementos são **maiores que 10** usando `while`.

````python
# Gabarito Exercício 17
nums = [5, 12, 8, 20, 3, 17]

i = 0
cont = 0

while i < len(nums):
    if nums[i] > 10:
        cont += 1
    i += 1

print(cont)
````

## Exercício 18 — Copiar lista com `while`
Dada a lista:

```python
nomes = ['Ana', 'Bruno', 'Carlos']
```

Crie uma nova lista com os mesmos elementos, usando `while`.

````python
# Gabarito Exercício 18
nomes = ['Ana', 'Bruno', 'Carlos']

copia = []
i = 0

while i < len(nomes):
    copia.append(nomes[i])
    i += 1

print(copia)
````

## Desafio extra — Estatísticas da lista

Dada uma lista de números, calcule em **um único loop**:

- soma
- média
- maior valor
- menor valor
- quantidade de números pares

````python
# Gabarito do desafio extra
nums = [10, 3, 8, 21, 14, 5]

soma = 0
maior = nums[0]
menor = nums[0]
qtd_pares = 0

for n in nums:
    soma += n

    if n > maior:
        maior = n

    if n < menor:
        menor = n

    if n % 2 == 0:
        qtd_pares += 1

media = soma / len(nums)

print('Soma:', soma)
print('Média:', media)
print('Maior:', maior)
print('Menor:', menor)
print('Quantidade de pares:', qtd_pares)

# Desvio padrão da lista
````
