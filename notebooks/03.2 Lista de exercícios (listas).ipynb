## 1) Lista `compras`
Crie `compras` com os itens iniciais, depois adicione `"biscoito"`, remova `"macarrão"` e mostre a lista final.

```python
# 1) compras
compras = ["arroz", "feijão", "macarrão", "frango", "suco"]

compras.append("biscoito")
compras.remove("macarrão")

compras
```

## 2) Indexação em `nums`
Dada `nums = [10, 20, 30, 40, 50, 60]`, mostre: primeiro, último, três primeiros e dois últimos.

```python
# 2) nums
nums = [10, 20, 30, 40, 50, 60]

primeiro = nums[0]
ultimo = nums[-1]
tres_primeiros = nums[:3]
dois_ultimos = nums[-2:]

primeiro, ultimo, tres_primeiros, dois_ultimos
```

## 3) Trocar palavra e reconstruir frase
Dada a frase `"python é legal demais"`, transforme em lista de palavras, troque `"legal"` por `"massa"` e reconstrua.

```python
# 3) frase -> lista -> troca -> frase
frase = "python é legal demais"

palavras = frase.split()
# palavras: ["python", "é", "legal", "demais"]

palavras[2] = "massa"  # troca "legal" por "massa"
frase2 = " ".join(palavras)

palavras, frase2
```

## 4) Formato "Sobrenome, Nome" e lista `pessoas`
Monte as strings e coloque tudo numa lista chamada `pessoas`.

```python
# 4) pessoas
nome1, sobrenome1 = "Ana", "Silva"
nome2, sobrenome2 = "Diego", "Souza"
nome3, sobrenome3 = "Marina", "Campos"

p1 = f"{sobrenome1}, {nome1}"
p2 = f"{sobrenome2}, {nome2}"
p3 = f"{sobrenome3}, {nome3}"

pessoas = [p1, p2, p3]
pessoas
```

## 5) Operações em `valores`
Dada `valores = [3, 1, 4, 1, 5]`: adicione 9, remova o **segundo** 1, crie `valores2 = valores + [6, 5]` e mostre ambos.

```python
# 5) valores e valores2
valores = [3, 1, 4, 1, 5]

valores.append(9)

# remover o segundo 1:
# (o primeiro 1 está no índice 1; o segundo 1 está no índice 3)
valores.pop(3)

valores2 = valores + [6, 5]

valores, valores2
```

## 6) Juntar palavras em frase com '!'
Dada `palavras = ["Hoje", "vai", "dar", "tudo", "certo"]`, junte em uma frase e coloque `!` ao final.

```python
# 6) frase com '!'
palavras = ["Hoje", "vai", "dar", "tudo", "certo"]

frase = " ".join(palavras) + "!"
frase
```

## 8) Formatar lista de animais
Dada a frase `"gato cachorro papagaio"`, crie `"gato, cachorro e papagaio"`.

## 9) Separar usuário e domínio do e-mail
Dado `"mariana.pereira@empresa.com.br"`, separe usuário e domínio e mostre: domínio completo, apenas `empresa`, e o final do domínio (`br`).

```python
# 9) e-mail
email = "mariana.pereira@empresa.com.br"

usuario, dominio = email.split("@")

dominio_completo = dominio                 # "empresa.com.br"
empresa = dominio.split(".")[0]            # "empresa"
final = dominio.split(".")[-1]             # "br"

usuario, dominio_completo, empresa, final
```
