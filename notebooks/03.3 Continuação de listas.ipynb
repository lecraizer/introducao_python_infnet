# Python — Listas (sem `if`, `for`, `while`)
Este notebook é um guia **prático** para treinar **listas**: criação, indexação, slicing, concatenação, `append`, `extend`, `insert`, `remove`, `pop`, `del`, `clear`, `count`, `index`, `sort`, `reverse`, `copy`, `len`, `min`, `max`, `sum`, e conversões com `list()`.

> **Regra do jogo (por enquanto):** nada de controle de fluxo (`if/else`, `for`, `while`) e nada de list comprehension.  
Tudo aqui dá pra resolver usando apenas **operações e métodos de lista**.


## 1) Criando listas e entendendo o que é um índice
- Uma lista guarda vários valores em uma única variável.
- Índices começam em `0`.
- Índice negativo conta de trás pra frente (`-1` é o último).


````python
# Criação
numeros = [10, 20, 30, 40]
nomes = ["Ana", "Bia", "Caio"]
mistura = [1, "oi", 3.5, True]

numeros, nomes, mistura
````

### Indexação básica
Experimente acessar itens específicos.

````python
numeros[0], numeros[1], numeros[-1], nomes[-2]
````

## 2) Slicing (fatiamento)
O slicing pega **pedaços** da lista.

Forma geral: `lista[inicio:fim:passo]`  
- `inicio` (inclui)  
- `fim` (não inclui)  
- `passo` (opcional)


````python
numeros[1:3], numeros[:2], numeros[2:], numeros[::2], numeros[::-1]
````

## 3) Concatenação e repetição de listas
- `+` cria uma **nova** lista.
- `*` repete elementos.


````python
a = [1, 2, 3]
b = [4, 5]
a + b
````

````python
b * 3
````

## 4) Adicionando itens: `append`, `extend`, `insert`
- `append(x)` adiciona **um** item no final (mesmo que seja uma lista).
- `extend(iterável)` adiciona **vários** itens (espalha).
- `insert(i, x)` insere em uma posição.


````python
lista = ["a", "b", "c"]
lista.append("d")
lista
````

````python
lista2 = ["a", "b", "c"]
lista2.append(["x", "y"])   # vira lista dentro de lista
lista2
````

````python
lista3 = ["a", "b", "c"]
lista3.extend(["x", "y"])   # adiciona cada item
lista3
````

````python
lista4 = ["a", "b", "c"]
lista4.insert(1, "Z")       # insere na posição 1
lista4
````

## 5) Removendo itens: `remove`, `pop`, `del`, `clear`
- `remove(x)` remove o **primeiro** item igual a `x`.
- `pop()` remove e **devolve** o último; `pop(i)` remove e devolve o item da posição `i`.
- `del lista[i]` remove por índice (ou por slice).
- `clear()` apaga tudo.


````python
itens = ["a", "b", "b", "c"]
itens.remove("b")  # remove só o primeiro "b"
itens
````

````python
nums = [10, 20, 30, 40]
removido = nums.pop()   # tira o último
nums, removido
````

````python
nums2 = [10, 20, 30, 40]
removido2 = nums2.pop(1)  # tira o índice 1
nums2, removido2
````

````python
nums3 = [10, 20, 30, 40, 50]
del nums3[1]      # remove o índice 1
nums3
````

````python
nums4 = [10, 20, 30, 40, 50]
del nums4[1:4]    # remove do índice 1 até 3
nums4
````

````python
apagavel = [1, 2, 3]
apagavel.clear()
apagavel
````

## 6) Medidas e buscas: `len`, `count`, `index`, operador `in`
- `len(lista)` = tamanho  
- `count(x)` = quantas vezes `x` aparece  
- `index(x)` = posição do primeiro `x`  
- `x in lista` = `True/False` se existe


````python
dados = ["uva", "maçã", "uva", "banana", "uva"]
len(dados), dados.count("uva"), dados.index("banana"), ("pera" in dados)
````

## 7) Ordenação e reversão: `sort`, `sorted`, `reverse`
- `sort()` altera a lista **no lugar**.
- `sorted(lista)` cria uma **nova** lista ordenada.
- `reverse()` inverte a lista **no lugar**.


````python
valores = [5, 1, 9, 2]
valores.sort()
valores
````

````python
valores2 = [5, 1, 9, 2]
ordenada = sorted(valores2)
valores2, ordenada
````

````python
valores3 = [1, 2, 3, 4]
valores3.reverse()
valores3
````

## 8) Cópia vs referência: cuidado!
Se você faz `b = a`, as duas variáveis apontam para a **mesma** lista.

Para copiar de verdade:
- `a.copy()`
- `a[:]`
- `list(a)`


````python
a = [1, 2, 3]
b = a
b.append(99)
a, b  # as duas mudam
````

````python
a = [1, 2, 3]
b = a.copy()
b.append(99)
a, b  # agora são independentes
````

## 9) Lista dentro de lista (matriz simples)
Você pode acessar usando dois índices: `matriz[linha][coluna]`.


````python
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
matriz[0], matriz[2][1], matriz[1][0:2]
````
